# Trainable Parameters of SimpleRNN

Model:

```python
model = Sequential()

model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

## Step 1: Parameters in `SimpleRNN`

A SimpleRNN has three trainable components:

1. Input-to-hidden weights ($W_x$)
2. Hidden-to-hidden (recurrent) weights ($W_h$)
3. Bias ($b$)

The total number of trainable parameters is

$$
\text{Parameters} = (I \times H) + (H \times H) + H
$$

or equivalently,

$$
\boxed{\text{Parameters} = H(I + H + 1)}
$$

where:

- $I$ = number of input features
- $H$ = number of hidden units

---

## Applying to this model

Input shape:

```python
input_shape = (50,1)
```

- Timesteps = **50**
- Input features = **1**

Number of hidden units:

```python
SimpleRNN(32)
```

Therefore,

$$
I = 1,\qquad H = 32
$$

### 1. Input-to-hidden weights

$$
I \times H = 1 \times 32 = 32
$$

### 2. Hidden-to-hidden (Recurrent) weights

$$
H \times H = 32 \times 32 = 1024
$$

### 3. Bias

One bias for each hidden unit:

$$
H = 32
$$

### Total parameters in SimpleRNN

$$
32 + 1024 + 32 = 1088
$$

or using the formula,

$$
\begin{aligned}
\text{Parameters}
&= H(I + H + 1) \\
&= 32(1 + 32 + 1) \\
&= 32 \times 34 \\
&= 1088
\end{aligned}
$$

# Why don't the 50 timesteps affect the parameter count?

The **same RNN cell is reused at every timestep**, meaning the weights are **shared** across all timesteps.

For example,

- Input shape `(50,1)` → **1088 parameters**
- Input shape `(100,1)` → **1088 parameters**
- Input shape `(500,1)` → **1088 parameters**

Only the **number of input features ($I$)** and the **number of hidden units ($H$)** determine the trainable parameters.

The sequence length (timesteps) affects **how many times the RNN cell is executed**, but **not** the number of trainable parameters.

---

# General Formula

For

```python
SimpleRNN(units=H, input_shape=(timesteps, I))
```

the number of trainable parameters is

$$
\boxed{\text{Parameters} = H(I + H + 1)}
$$

where

- $I$ = input features
- $H$ = hidden units

The number of timesteps does **not** appear in the formula because the same weights are reused at every timestep.
