```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN
```

```python
model=Sequential()

model.add(SimpleRNN(3,input_shape=(4,5))) # we assume maximum size of input will be 4 words and 5 features(vocab words), Also for RNN,by default the activation function will be tanh
model.add(Dense(1,activation="sigmoid"))

model.summary()
```

# firs layer weights

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```

## feedback connection weights within hidden layer itself

```python
print(model.get_weights()[1].shape)
model.get_weights()[1]
```

# biases for that hidden layer

```python
print(model.get_weights()[2].shape)
model.get_weights()[2]
```

# for weights betn hidden layer and output layer

```python
print(model.get_weights()[3].shape)
model.get_weights()[3]
```

# weights for output layer

```python
print(model.get_weights()[4].shape)
model.get_weights()[4]
```
