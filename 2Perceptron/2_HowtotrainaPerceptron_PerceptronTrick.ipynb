
## Topic 1: Introduction to Training a Perceptron

### 1. Introduction

**What is this topic?**  
This topic explains how a **perceptron** (a simple type of artificial neuron) learns to make correct predictions. In the previous lecture, you learned what a perceptron is and how it makes a prediction. But you did not learn how to find the correct values for its **weights** and **bias** (called “rates and biceps” in the transcript – this is a verbal slip, meaning **weights and biases**).  
This lecture fills that gap: it teaches you how to **train** a perceptron so it can classify data correctly.

**Why is it important?**  
Without training, a perceptron makes random or wrong predictions. Training is the process that adjusts the perceptron’s internal numbers so it learns from examples. This is the foundation of almost all modern machine learning and deep learning.

**Real-life use:**  
- Classifying emails as “spam” or “not spam”  
- Deciding if a student will get placed based on marks  
- Teaching a computer to separate two types of objects in an image  

---

### 2. Detailed Explanation

#### What does “linearly separable” mean?

The transcript says: *“this data is linearly separable, which means you can classify this data by drawing a line into these two classes.”*

Imagine you have two types of points on a paper:  
- Green points (one class, e.g., “placed”)  
- Blue points (another class, e.g., “not placed”)  

If you can draw a **single straight line** that separates all green points on one side and all blue points on the other side, the data is **linearly separable**.

> **Analogy:** Think of green and blue candies mixed on a table. If you can place a ruler so that all green candies are on one side and all blue on the other, the candies are linearly separable.

#### What is the “line” in a perceptron?

A perceptron’s decision is based on a straight line (in 2D) or a flat surface (in 3D) or a hyperplane (in higher dimensions).  
The line is described by an equation:

```
w1 * x1 + w2 * x2 + b = 0
```

Where:  
- `x1`, `x2` are the input features (e.g., marks in two subjects)  
- `w1`, `w2` are **weights** (importance of each feature)  
- `b` is the **bias** (intercept, called “not” in the transcript)

The line divides the space into two regions:  
- **Positive region:** where `w1*x1 + w2*x2 + b > 0` → prediction = 1 (e.g., placed)  
- **Negative region:** where `w1*x1 + w2*x2 + b < 0` → prediction = 0 (e.g., not placed)

> The transcript says: *“if this number is positive, then the prediction will be 1. If it is negative, then the prediction will be zero.”*

#### What is the problem with a bad line?

Initially, the perceptron starts with random weights and bias. The line will be in the wrong place.  
Some points will be on the wrong side:  
- A **blue point** (actual class = 0) in the **positive region** (prediction = 1) → this is a **misclassified point**  

The transcript calls this: *“a negative point which is stuck in the positive region”* or *“positive point in the negative region.”*

To fix this, the perceptron must **move the line** toward the misclassified point.

---

### 3. Key Points

- **Training** = finding correct weights and bias for a perceptron.
- **Linearly separable data** = data that can be separated by a single straight line.
- **Positive region** = side of the line where prediction is 1.
- **Negative region** = side of the line where prediction is 0.
- **Misclassified point** = a point that lies on the wrong side of the line.
- The goal of training is to move the line so that all points are on the correct side.

## Topic 2: Positive and Negative Regions – How to Identify Them

### 1. Introduction

**What is this topic?**  
This topic explains how to determine which side of a line is the **positive region** (prediction = 1) and which side is the **negative region** (prediction = 0). The transcript emphasizes: *“how will you know which region will be green and which region will be blue”* — meaning, how do you decide the class for each side?

**Why is it important?**  
Without knowing which side means what, you cannot make predictions. The perceptron’s entire decision depends on whether a point falls in the positive or negative region. This knowledge is also essential for training, because when you see a misclassified point, you need to know which region it is *actually in* versus which region it *should be in*.

**Real-life use:**  
- Medical diagnosis: positive region = “disease present”, negative region = “disease absent”  
- Loan approval: positive region = “approve loan”, negative region = “reject loan”  
- Any binary classification (yes/no decision)

---

### 2. Detailed Explanation

#### What is the equation of a line?

In the transcript, the instructor gives an example:

```
2x + 3y + 5 = 0
```

This is a straight line.  
- `x` and `y` are coordinates (like input features)  
- `2`, `3`, and `5` are constants

> **Analogy:** Think of this line as a fence. Everything on one side of the fence is called “positive region,” and everything on the other side is “negative region.”

#### How do you find the positive region?

The transcript states: *“If you have any point, and you want to find out if any point is on the positive side of this line, what do you have to do? You know the equation of the line, equal to zero.”*

**Step-by-step method:**

1. Take the line equation and write it as: `2x + 3y + 5 = 0`
2. To test a point `(x, y)`, plug the `x` and `y` values into the left-hand side: `2x + 3y + 5`
3. Do not set it equal to zero. Instead, just calculate the number.
4. **Rule:**
   - If `2x + 3y + 5 > 0` → the point is in the **positive region**
   - If `2x + 3y + 5 < 0` → the point is in the **negative region**

> The transcript confirms: *“3y + 5 greater than zero. This entire region is displayed in blue. This is the positive region.”*

#### Example: Testing a point

Line: `2x + 3y + 5 = 0`

Test point A: `(0, 0)`
- Calculate: `2(0) + 3(0) + 5 = 5`
- Is 5 > 0? Yes.
- Conclusion: `(0, 0)` is in the **positive region**.

Test point B: `(-3, -1)`
- Calculate: `2(-3) + 3(-1) + 5 = -6 - 3 + 5 = -4`
- Is -4 > 0? No, it is less than 0.
- Conclusion: `(-3, -1)` is in the **negative region**.

#### What if the line equation is different?

The transcript gives another example: `-2x + 3y - 5 = 0`

The same rule applies:
- Calculate `-2x + 3y - 5`
- If result > 0 → positive region
- If result < 0 → negative region

> **Important:** The positive region is *not* always “above” the line. It depends on the equation. You must always test with a point.

#### How to know which region is which without testing every time?

Once you test **one point** on one side, you know the entire side.  
Example: Test `(0, 0)`. If it is positive, then the entire side containing `(0, 0)` is the positive region.

---

### 3. Key Points

- The line equation is set to **zero**: `w1*x1 + w2*x2 + b = 0`
- To check a point, **do not set to zero** — just calculate the expression.
- If **result > 0** → positive region (predict 1)
- If **result < 0** → negative region (predict 0)
- The positive region is not always “up” or “right” — it depends on the numbers.
- Test one known point to determine which side is positive.

---

### 4. Syntax/Structure (Not applicable – conceptual topic)

But here is the **decision rule** in simple form:

```
given: line: w1*x1 + w2*x2 + b = 0
given: a point (x1, x2)

compute: value = w1*x1 + w2*x2 + b

if value > 0:
    region = "positive" → prediction = 1
else if value < 0:
    region = "negative" → prediction = 0
```

---

### 5. Code Examples

The transcript does not provide explicit code for this section, but based on the explanation, here is a simple Python example:

- 5 > 0, so positive region.

## Topic 3: Three Types of Line Transformations (Moving and Rotating the Line)

### 1. Introduction

**What is this topic?**  
This topic explains how a line can be **moved** or **rotated** by changing its coefficients (weights and bias). The transcript describes three basic transformations: changing the bias (c) to move the line up/down, and changing the weights to rotate the line. These transformations are the foundation of how a perceptron learns to correct misclassified points.

**Why is it important?**  
When a point is misclassified, the perceptron needs to adjust the line to bring it closer to that point. To do that, you must understand how each coefficient affects the line's position and angle. Without this knowledge, you cannot control the learning process.

**Real-life use:**  
- Adjusting a spam filter: moving the decision boundary to catch more spam emails  
- Tuning a credit score model: rotating the boundary to better separate good vs. bad borrowers  
- Any system where a line (or plane) must be shifted to improve classification

---

### 2. Detailed Explanation

The transcript mentions: *“you can do three different types of transformations”* by changing coefficients.

#### General form of a line

The perceptron line (in 2D) is written as:

```
w1 * x1 + w2 * x2 + b = 0
```

Where:
- `w1` = weight for feature 1 (coefficient of x1)
- `w2` = weight for feature 2 (coefficient of x2)
- `b` = bias (also called intercept, or "c" in the transcript)

> The transcript writes it as: `2x + 3y + 5 = 0` → here, w1 = 2, w2 = 3, b = 5

#### Transformation 1: Changing the bias (b) — Parallel movement

The transcript says: *“if you change c, if you increase or decrease c, your line goes up or down parallel.”*

**What happens?**
- Increasing `b` → line moves in one direction (down, according to the transcript's example)
- Decreasing `b` → line moves in the opposite direction (up)

**Example from transcript:**
```
2x + 3y + 5 = 0   (original line)
2x + 3y + 0 = 0   (after decreasing c from 5 to 0)
```
The transcript notes: *“3y + 0 equals zero. So this purple line goes upwards.”*

> **Analogy:** Imagine a straight fence (the line). Changing `b` is like sliding the entire fence left/right or up/down without changing its angle. It remains parallel to itself.

#### Transformation 2: Changing w1 (coefficient of x) — Rotation

The transcript states: *“2x + a + 5 = 0. So this is a rotation.”*

**What happens?**
- Changing `w1` rotates the line around its intersection point.
- The line's slope changes.

> **Analogy:** Imagine a door hinge. Changing `w1` is like pushing the door to a different angle. The line pivots.

#### Transformation 3: Changing w2 (coefficient of y) — Rotation

Similarly, changing `w2` also rotates the line. Together, `w1` and `w2` control the angle (slope) of the line.

#### Combining transformations

The transcript summarizes: *“if you apply these three things individually, then transformation occurs in three different types of effect combinations.”*

By changing `w1`, `w2`, and `b` together, you can move the line to **any position and angle** — which is exactly what training does.

#### How transformations help training

When a point is misclassified, the perceptron:
1. Identifies which side the point is on
2. Adjusts `w1`, `w2`, and `b` to move the line **toward** that point

The transcript gives a specific rule:
- *“If you want to move this line in this direction, then what does it say, you will add a five one at the end of this coordinate.”*
- *“When you want to move towards the positive region, you'll do minus.”*
- *“When you want to move towards the negative region, you'll do plus.”*

> Don't worry — the exact math for this will be covered in the next topic (Perceptron Trick).

---

### 3. Key Points

| Transformation | What to change | Effect on line |
|----------------|----------------|----------------|
| 1 | Bias (b) | Parallel movement (up/down or left/right) |
| 2 | Weight w1 | Rotation |
| 3 | Weight w2 | Rotation |

- Changing `b` alone keeps the line **parallel** to itself.
- Changing `w1` or `w2` changes the **slope** (angle).
- Training uses **combinations** of all three to move the line toward misclassified points.
- Direction of movement (plus or minus) depends on whether the point is in the positive or negative region.

---

### 4. Syntax/Structure (Not applicable — conceptual)

But here is a simple way to remember:

```
Line: w1 * x + w2 * y + b = 0

Increase b → line moves one way
Decrease b → line moves opposite way

Increase w1 → line rotates one direction
Decrease w1 → line rotates opposite direction

Same for w2
```

---

### 5. Code Examples

The transcript does not provide code for transformations alone, but here is a demonstration using Python to visualize how changing `b` moves the line parallel:

```python
# Simple demonstration of parallel movement by changing bias (b)

def line_value(x, y, w1, w2, b):
    return w1 * x + w2 * y + b

# Fixed weights
w1 = 2
w2 = 3

# Different bias values
biases = [5, 0, -5]

print("Testing point (0,0) with different biases:")
for b in biases:
    result = line_value(0, 0, w1, w2, b)
    print(f"b = {b}: result = {result} → {'positive region' if result > 0 else 'negative region' if result < 0 else 'on line'}")
```

**Expected output:**

**Explanation:**  
Changing `b` shifted the line so that the same point `(0,0)` moved from positive region (b=5) to exactly on the line (b=0) to negative region (b=-5). The line's angle did not change — only its position.

**Q1:** What is the effect of increasing the bias (b) in the line equation w1*x1 + w2*x2 + b = 0?  
**A1:** It shifts the line parallel to itself. The direction (up/down, left/right) depends on the specific equation, but the angle (slope) does not change.

**Q2:** How do you rotate the decision boundary of a perceptron?  
**A2:** By changing the weights w1 and/or w2. Changing the bias (b) only moves the line parallel, without rotation.

**Q3:** If a misclassified point is above the line, can you fix it by changing only the bias?

  2. Change `w1` → rotation

- **Training** = combining these transformations to move line toward misclassified points  
- **Parallel movement** = changing position, not angle  
- **Rotation** = changing angle, not just position

## Topic 4: The Perceptron Trick – Moving the Line Toward a Misclassified Point

### 1. Introduction

**What is this topic?**  
This topic explains the **Perceptron Trick** — the mathematical rule that tells you exactly how to change the weights and bias to move the line toward a misclassified point. The transcript describes this as: *“how this transformation is happening, that is, if I am taken towards the point, how can I take my line.”*

**Why is it important?**  
This is the **core learning rule** of the perceptron. Without this trick, the perceptron cannot learn. Every time the perceptron makes a mistake, this rule tells it how to adjust itself to do better next time.

**Real-life use:**  
- Every time you train a simple neural network  
- Online learning systems that update instantly when a mistake is made  
- Adaptive spam filters that improve as you mark emails as spam/not spam  

---

### 2. Detailed Explanation

#### The problem setup

From the transcript:  
- You have a line (decision boundary): `w1*x1 + w2*x2 + b = 0`  
- You have a point that is **misclassified**  
- You want to move the line **toward** that point so that eventually the point ends up on the correct side

#### What happens during misclassification?

The transcript gives two cases:

**Case 1:** *“negative point in the positive region”*  
- Actual class = 0 (negative)  
- Predicted class = 1 (positive, because it's in positive region)  
- This point should NOT be here

**Case 2:** *“positive point in the negative region”*  
- Actual class = 1 (positive)  
- Predicted class = 0 (negative)  
- This point should NOT be here either

#### The transformation rule (without learning rate first)

The transcript explains the basic idea:

> *“If a negative point is in the positive region, you add a 1 to its coordinates and get the coefficient of the line.”*

And later:  
> *“When you want to move towards the negative region, you'll do plus.”*  
> *“When you want to move towards the positive region, you'll do minus.”*

But careful — the transcript clarifies the actual math:

**For a negative point in the positive region (misclassified):**
- You need to move the line toward the negative region
- You **add** something to the coefficients

**For a positive point in the negative region (misclassified):**
- You need to move the line toward the positive region  
- You **subtract** something from the coefficients

#### The problem with big jumps

The transcript warns:  
> *“You don't apply a large transformation in a single step. You never apply such a large transformation.”*

If you move the line too much in one step, you might overshoot or break the learning. Instead, you take **small, careful steps**.

#### Introducing the learning rate

> *“Instead, what you do is you take a small number, which we call the learning rate.”*

The learning rate (often written as `η` or `lr` or `alpha`) is a **small positive number**, like `0.01` or `0.1`. It controls how big each step is.

#### The complete Perceptron Trick formula

The transcript gives the formula:

> *“First you combine all the coordinates with the learning rate. You have to multiply and then subtract from the old coefficient, then you get the new coefficient.”*

**For updating weights (w1, w2):**

```
new_w1 = old_w1 + (learning_rate) * (actual_class - predicted_class) * x1
new_w2 = old_w2 + (learning_rate) * (actual_class - predicted_class) * x2
new_b   = old_b   + (learning_rate) * (actual_class - predicted_class) * 1
```

Wait — the transcript mentions subtraction. Let me explain both ways:

From the transcript: *“old minus will actually become a plus”* — this depends on how you define things.

The standard Perceptron Trick (simplified from the transcript):

**If point is misclassified:**
```
new_w1 = old_w1 + learning_rate * x1 * (desired correction direction)
new_w2 = old_w2 + learning_rate * x2 * (desired correction direction)
new_b  = old_b  + learning_rate * 1 * (desired correction direction)
```

Where correction direction is:
- `+1` if we need to move toward positive region
- `-1` if we need to move toward negative region

The transcript simplifies this later by using a **single rule without conditions** (which we will see in Topic 5).

---

#### Step-by-step example

Suppose:
- Current line: `2x + 3y + 5 = 0`  (w1=2, w2=3, b=5)
- Misclassified point: `(x1=1, x2=1)`, actual class = 0 (negative)
- But it's in positive region (because `2*1+3*1+5=10 > 0`)
- Learning rate = 0.1

**Step 1:** We need to move line toward negative region (so subtract)
**Step 2:** Update:
```
new_w1 = 2 - 0.1 * 1 = 1.9
new_w2 = 3 - 0.1 * 1 = 2.9
new_b  = 5 - 0.1 * 1 = 4.9
```
**Step 3:** New line: `1.9x + 2.9y + 4.9 = 0`
**Step 4:** Test the same point again: `1.9*1+2.9*1+4.9 = 9.7` — still positive, but slightly smaller. We need many steps.

---

### 3. Key Points

- **Perceptron Trick** = the rule for updating weights and bias when a point is misclassified
- **Learning rate** = a small number that controls step size (prevents big jumps)
- Without learning rate, changes are too large and learning fails
- Formula (basic form):  
  `new_weight = old_weight + learning_rate * input * direction`
- Direction depends on whether point should move toward positive or negative region
- You never fix misclassification in **one step** — you take many small steps

---

### 4. Syntax/Structure

**General update rule (conceptual):**

```
for each misclassified point (x1, x2) with actual class y (0 or 1):
    if point is in positive region but y == 0:  # negative point in positive region
        w1 = w1 - learning_rate * x1
        w2 = w2 - learning_rate * x2
        b  = b  - learning_rate * 1
    else if point is in negative region but y == 1:  # positive point in negative region
        w1 = w1 + learning_rate * x1
        w2 = w2 + learning_rate * x2
        b  = b  + learning_rate * 1
```

**Where:**
- `w1, w2` = weights
- `b` = bias
- `learning_rate` = small number (e.g., 0.01)
- `x1, x2` = input features of misclassified point
- The `1` for bias update comes from the fact that bias is like a weight for a constant input of 1

---

### 5. Code Examples

The transcript does not give full code for this section, but here is a simple implementation of the Perceptron Trick:

```python
# Simple Perceptron Trick for one misclassified point

# Current weights and bias
w1 = 2.0
w2 = 3.0
b = 5.0

# Misclassified point
x1 = 1.0
x2 = 1.0
actual_class = 0  # 0 = negative (blue), 1 = positive (green)

# Learning rate
learning_rate = 0.1

# Check which region the point is in
prediction_value = w1 * x1 + w2 * x2 + b
print(f"Current prediction value: {prediction_value}")

if prediction_value > 0:
    print("Point is in POSITIVE region")
    region = "positive"
else:
    print("Point is in NEGATIVE region")
    region = "negative"

# Apply Perceptron Trick
if region == "positive" and actual_class == 0:
    # Negative point in positive region → move line toward negative region (subtract)
    print("Case: Negative point in positive region → SUBTRACT")
    w1 = w1 - learning_rate * x1
    w2 = w2 - learning_rate * x2
    b = b - learning_rate * 1
elif region == "negative" and actual_class == 1:
    # Positive point in negative region → move line toward positive region (add)
    print("Case: Positive point in negative region → ADD")
    w1 = w1 + learning_rate * x1
    w2 = w2 + learning_rate * x2
    b = b + learning_rate * 1
else:
    print("Point is correctly classified → no update")

print(f"New weights: w1={w1}, w2={w2}, b={b}")
```

**Expected output:**
```
Current prediction value: 10.0
Point is in POSITIVE region
Case: Negative point in positive region → SUBTRACT
New weights: w1=1.9, w2=2.9, b=4.9
```

**Explanation:**
- The point (1,1) gave prediction value 10 (>0) → positive region
- But actual class is 0 (negative) → misclassified
- We subtract to move line toward negative region
- New weights and bias are slightly smaller

## Topic 5: The Simplified Perceptron Training Algorithm (No If-Else)

### 1. Introduction

**What is this topic?**  
This topic presents a **simplified version** of the Perceptron Trick that does not require checking two separate conditions (positive point in negative region OR negative point in positive region). Instead, the transcript shows a **single, unconditional update rule** applied to every randomly selected point — whether it is misclassified or not. The transcript says: *“Rather than checking these two conditions, I will simply run a loop... every time, without checking anything, I will apply this logic. No if here.”*

**Why is it important?**  
This simplification makes the algorithm:
- **Easier to code** (no complex conditionals)
- **Efficient** (same rule for all points)
- **Mathematically elegant** (works because correctly classified points result in zero update)

**Real-life use:**  
This simplified approach is closer to how modern neural networks update weights using **gradient descent**. It forms the foundation for more advanced algorithms.

---

### 2. Detailed Explanation

#### The problem with two conditions

The original Perceptron Trick had two cases:
1. Negative point in positive region → subtract
2. Positive point in negative region → add

Writing code for this requires:
```python
if (prediction == 1 and actual == 0):
    # subtract
elif (prediction == 0 and actual == 1):
    # add
```

The transcript simplifies this by noting that both cases can be captured with **one mathematical expression**.

#### The key insight

The transcript explains by walking through two scenarios:

**Scenario 1: Student got placed (actual = 1), model said not placed (prediction = 0)**
- This is a **positive point in negative region**
- The transcript says: *“This is equal to w plus learning rate times in the positive region”*
- Update rule: `new_w = old_w + learning_rate * input`

**Scenario 2: Student did not get placed (actual = 0), model said placed (prediction = 1)**
- This is a **negative point in positive region**
- The transcript says: *“This is equal to w minus learning rate”*
- Update rule: `new_w = old_w - learning_rate * input`

#### The single unified rule

The transcript reveals the simplification:

> *“We will just apply this single rule.”*

And the rule is based on the **difference between actual and predicted**:

```
error = actual_class - predicted_class
```

- If actual = 1, predicted = 0 → error = +1 → add
- If actual = 0, predicted = 1 → error = -1 → subtract
- If actual = predicted → error = 0 → no change

#### The complete simplified update formula

```
new_w1 = old_w1 + learning_rate * (actual - predicted) * x1
new_w2 = old_w2 + learning_rate * (actual - predicted) * x2
new_b  = old_b  + learning_rate * (actual - predicted) * 1
```

**Why this works:**
- `(actual - predicted)` gives:
  - `+1` when actual=1 and predicted=0 (need to add)
  - `-1` when actual=0 and predicted=1 (need to subtract)
  - `0` when correct (no change)
- The learning rate scales the step size
- Multiplying by input features ensures each weight is adjusted proportionally to how much that feature contributed

#### The algorithm structure

From the transcript:

> *“I will write an algorithm... for i in the range of epoch (1000 times). Inside this loop, I will randomly select a point. Update my coefficients using this rule.”*

**Algorithm in plain English:**

```
1. Initialize weights (w1, w2) and bias (b) to small random numbers
2. Repeat 1000 times:
      Randomly pick one data point (x1, x2) with its actual class (y)
      Calculate prediction: 
          value = w1*x1 + w2*x2 + b
          predicted = 1 if value >= 0 else 0
      Calculate error = actual - predicted
      Update:
          w1 = w1 + learning_rate * error * x1
          w2 = w2 + learning_rate * error * x2
          b  = b  + learning_rate * error * 1
3. Return the final weights and bias
```

> The transcript emphasizes: *“No if here. I will directly update it.”*

**A1:** If all weights start at zero, the dot product will be zero for all inputs. The perceptron may not break symmetry effectively, leading to slower or failed learning. Small random values help the learning process.

**Q2:** In the weights array, why is bias stored as the first element?  
**A2:** It is a convention. The bias is like a weight for a constant input of 1. Storing it first or last doesn't matter as long as you are consistent. The transcript stores it as the first item.

**Q3:** What is the purpose of the step function?  
**A3:** The step function converts the continuous dot product (which can be any number) into a discrete class label (0 or 1). Without it, the output would not be a binary prediction.

**Q4:** How do you choose the number of epochs (1000 in the example)?  
**A4:** It is a hyperparameter. Too few epochs may not allow the perceptron to learn; too many epochs waste computation. In practice, you can train until convergence (no errors) or set a fixed large number.

**A5:** Updates become `weight += 0 * error * input = 0`. The weights never change, and the perceptron never learns.

#### The Complete Perceptron Training Algorithm

Based on the transcript, here is the **complete step-by-step algorithm**:

**Step 1: Understand the problem**
- Data must be **linearly separable** (a straight line can separate the two classes)
- Each data point has features `(x1, x2)` and an actual class `y` (0 or 1)

**Step 2: Define the model**
- Decision boundary (line): `w1 * x1 + w2 * x2 + b = 0`
- Prediction rule: 
  - If `w1*x1 + w2*x2 + b >= 0` → predict `1` (positive region)
  - If `w1*x1 + w2*x2 + b < 0` → predict `0` (negative region)

**Step 3: Initialize parameters**
- Set `w1`, `w2`, `b` to small random values (e.g., between -0.5 and 0.5)
- Set **learning rate** (e.g., `0.01` or `0.1`) — a small number controlling step size
- Set **number of epochs** (e.g., `1000`) — how many times to iterate

**Step 4: Training loop** (repeat for each epoch)
```
For epoch in range(epochs):
    1. Randomly select one training point (x1, x2, actual_class)
    2. Calculate: value = w1*x1 + w2*x2 + b
    3. Predict: predicted = 1 if value >= 0 else 0
    4. Calculate error = actual_class - predicted
    5. Update weights and bias:
        w1 = w1 + learning_rate * error * x1
        w2 = w2 + learning_rate * error * x2
        b  = b  + learning_rate * error * 1
```

**Step 5: Convergence**
- After enough epochs, the line will correctly classify all training points
- The algorithm stops (or you stop after fixed epochs)

```python
"""
Complete Perceptron Implementation
Based on Lecture 5: Training the Perceptron
"""

import random

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=1000):
        """
        Initialize the perceptron with hyperparameters.
        
        learning_rate: step size for weight updates (small number)
        epochs: number of training iterations
        """
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None  # [bias, w1, w2]
    
    def _initialize_weights(self, n_features):
        """Initialize weights to small random values."""
        # The transcript: "I basically created a weights array, with all three items initially"
        self.weights = [random.uniform(-0.5, 0.5) for _ in range(n_features + 1)]
        # weights[0] = bias, weights[1] = w1, weights[2] = w2
    
    def _predict_one(self, x1, x2):
        """Predict class for a single point (0 or 1)."""
        # The transcript: "dot product and this is the answer I am getting"
        value = self.weights[1] * x1 + self.weights[2] * x2 + self.weights[0]
        # The transcript: "if I write step 0.294 then prediction is 0"
        return 1 if value >= 0 else 0
    
    def train(self, X, y):
        """
        Train the perceptron on labeled data.
        
        X: list of [x1, x2] points
        y: list of actual classes (0 or 1)
        """
        # Step 1: Initialize
        n_features = len(X[0])  # Should be 2 for our data
        self._initialize_weights(n_features)
        
        print(f"Initial weights: bias={self.weights[0]:.4f}, w1={self.weights[1]:.4f}, w2={self.weights[2]:.4f}")
        print(f"Training for {self.epochs} epochs with learning rate {self.lr}\n")
        
        # Step 2: Training loop
        # The transcript: "for i in the range of epoch value, thousand"
        for epoch in range(self.epochs):
            # Randomly select a point
            # The transcript: "every time inside the loop, I selected a random point"
            idx = random.randint(0, len(X) - 1)
            x1, x2 = X[idx]
            actual = y[idx]
            
            # Predict
            predicted = self._predict_one(x1, x2)
            
            # Calculate error
            # The transcript: "error = actual - predicted"
            error = actual - predicted
            
            # Update weights and bias (simplified rule, no if-else)
            # The transcript: "weights = weights + learning_rate * error * [1, x1, x2]"
            self.weights[0] += self.lr * error * 1   # bias update
            self.weights[1] += self.lr * error * x1  # w1 update
            self.weights[2] += self.lr * error * x2  # w2 update
            
            # Optional: print progress occasionally
            if epoch % 200 == 0 and epoch > 0:
                print(f"Epoch {epoch}: bias={self.weights[0]:.4f}, w1={self.weights[1]:.4f}, w2={self.weights[2]:.4f}")
        
        print(f"\nTraining complete! Final weights: bias={self.weights[0]:.4f}, w1={self.weights[1]:.4f}, w2={self.weights[2]:.4f}")
    
    def predict(self, X):
        """Predict classes for multiple points."""
        return [self._predict_one(x1, x2) for x1, x2 in X]
    
    def evaluate(self, X, y):
        """Calculate accuracy on test data."""
        predictions = self.predict(X)
        correct = sum(1 for pred, actual in zip(predictions, y) if pred == actual)
        accuracy = correct / len(y) * 100
        return accuracy, predictions


# ============ DEMONSTRATION ============

if __name__ == "__main__":
    # Training data (linearly separable)
    # Features: [marks_in_math, marks_in_english]
    # Output: 1 = placed, 0 = not placed
    X_train = [
        [8, 7],   # high marks → placed
        [3, 4],   # low marks → not placed
        [9, 6],   # high marks → placed
        [2, 3],   # low marks → not placed
        [7, 8],   # high marks → placed
        [4, 2]    # low marks → not placed
    ]
    y_train = [1, 0, 1, 0, 1, 0]
    
    # Create and train perceptron
    perceptron = Perceptron(learning_rate=0.1, epochs=1000)
    perceptron.train(X_train, y_train)
    
    # Evaluate on training data
    accuracy, predictions = perceptron.evaluate(X_train, y_train)
    
    print("\n" + "="*50)
    print("FINAL RESULTS")
    print("="*50)
    for i, ((x1, x2), actual, pred) in enumerate(zip(X_train, y_train, predictions)):
        status = "✓" if pred == actual else "✗"
        print(f"Point {i+1}: ({x1},{x2}) actual={actual} predicted={pred} {status}")
    
    print(f"\nAccuracy on training data: {accuracy:.1f}%")
    
    # Test a new point
    test_point = [6, 6]
    test_pred = perceptron._predict_one(test_point[0], test_point[1])
    print(f"\nTest on new student with marks ({test_point[0]},{test_point[1]}): predicted = {test_pred} ({'Placed' if test_pred==1 else 'Not placed'})")
```

**Q2:** What are the three hyperparameters in the perceptron training algorithm described in the transcript?

**Q3:** Why does the simplified algorithm not need if-else conditions for misclassification?  
**A3:** Because `error = actual - predicted` automatically gives +1 (needs addition), -1 (needs subtraction), or 0 (correct). The same formula handles all cases.

**Q4:** What does the transcript mean by "convergence"?  
**A4:** Convergence means the perceptron has found a line that correctly classifies all training points. After convergence, the line stops moving because there are no errors.

**Q5:** Can the perceptron learn if the data is not linearly separable?  
**A5:** No. The perceptron algorithm only converges (finds a solution) if the data is linearly separable. For non-separable data, the line will keep moving forever without finding a perfect solution.

**Q6:** What is the purpose of the learning rate in the update rule?  
**A6:** The learning rate controls how much the line moves in each step. A small learning rate ensures stable, gradual learning. A large learning rate may cause overshooting or failure to converge.

**Q7:** How do you know when to stop training?  
**A7:** Stop when the perceptron correctly classifies all training points (convergence) or after a fixed number of epochs (if convergence is not reached).
