## Topic 1: Review of the Perceptron (Background Knowledge)

### 1. Introduction

**What is this topic?**  
This topic reviews the basic perceptron model, which is a foundational concept in machine learning. A perceptron is a simple mathematical model inspired by a biological neuron in the human brain.

**Why is it important?**  
Before learning about loss functions, you must understand how a perceptron works. The perceptron trick (a simple training method) has problems, and loss functions are the solution to those problems.

**Where is it used in real life?**  
Perceptrons are used for binary classification tasks, such as:
- Deciding whether a student will get placed based on CGPA and IQ
- Classifying emails as spam or not spam
- Determining if a loan applicant will default or not

---

### 2. Detailed Explanation

#### What is a Perceptron?

A perceptron is a mathematical model that:
- Takes multiple inputs (like CGPA, IQ, etc.)
- Multiplies each input by a **weight**
- Adds all these products together plus a **bias** (also called intercept)
- Passes the result through a **step function** to produce an output (usually 1 or -1, or 1 or 0)

#### The Structure of a Perceptron

```
Input 1 (x₁) ----→ [multiply by weight w₁] ----+
Input 2 (x₂) ----→ [multiply by weight w₂] ----+----→ [Sum] ----→ [Step Function] ----→ Output
                                              ( + bias b )
```

**Step-by-step process:**

1. **Inputs**: You have data, for example:
   - x₁ = CGPA of a student
   - x₂ = IQ of a student

2. **Weights (w₁, w₂)**: Each input has a weight. Weights determine how important each input is. Initially, weights are random.

3. **Bias (b)**: An extra number added to the sum. It allows the line to shift away from the origin.

4. **Sum (Z)**: Calculate Z = (w₁ × x₁) + (w₂ × x₂) + b

5. **Step Function (Activation Function)**:  
   - If Z ≥ 0, output = 1 (positive class, e.g., "placed")
   - If Z < 0, output = -1 or 0 (negative class, e.g., "not placed")

#### Geometric Intuition of the Perceptron

- A perceptron draws a **straight line** (in 2D) or a **plane** (in 3D) or a **hyperplane** (in higher dimensions) to separate two classes of points.
- The line divides the space into two regions:
  - **Positive side**: Output = 1
  - **Negative side**: Output = -1
- This makes the perceptron a **binary classifier** (it only separates things into two groups).

#### Two Phases of a Machine Learning Algorithm

| Phase | What it does |
|-------|---------------|
| **Prediction** | Using existing weights and bias to calculate output for new data |
| **Training** | Calculating the best weights and bias using data |

#### The Perceptron Trick (Review)

This is a simple training method:
1. Start with a random line (random weights and bias)
2. Pick a random point
3. Check if the point is on the correct side of the line
4. If it is **misclassified**, pull the line toward that point
5. If it is **correctly classified**, do nothing
6. Repeat thousands of times

After many iterations, the line should settle in a good position.

---

### 3. Key Points

- A perceptron is a mathematical model inspired by a biological neuron.
- It takes inputs, multiplies by weights, adds a bias, and passes through a step function.
- Output is either 1 or -1 (binary classification).
- Geometrically, a perceptron is a line (or plane) that separates two classes.
- Training finds the best weights and bias.
- The perceptron trick moves the line only when a point is misclassified.

## Topic 2: Problems with the Perceptron Trick

### 1. Introduction

**What is this topic?**  
This topic explains the limitations and flaws of the perceptron trick (the simple training method where you pull the line toward misclassified points).

**Why is it important?**  
Understanding these problems motivates the need for loss functions. The perceptron trick works most of the time, but it has fundamental issues that prevent it from being a perfect solution.

**Where is it used in real life?**  
When building real machine learning systems, you cannot rely on the perceptron trick because:
- You need to know if your model is the best possible
- You need to guarantee your model will train successfully
- You need to compare different models objectively

---

### 2. Detailed Explanation

#### How the Perceptron Trick Works (Review)

The perceptron trick follows this simple logic:

| Situation | Action |
|-----------|--------|
| Point is **misclassified** | Pull the line toward that point (change weights and bias) |
| Point is **correctly classified** | Do nothing (no change to weights or bias) |

This seems reasonable, but it has two major problems.

---

#### Problem 1: Cannot Tell Which Line is Best

**The Issue:**

- Multiple different lines can correctly classify all points
- The perceptron trick randomly picks points to adjust the line
- Depending on which random points you pick, you can end up with different final lines
- The trick gives you **no way to know** which line is better

**Example from the transcript:**

Imagine you have a dataset. After running the perceptron trick, you might get:
- **Line A** that separates the classes
- **Line B** that also separates the classes

Both lines work perfectly (they classify all points correctly). But which one is better? The perceptron trick cannot tell you.

**Why this is a problem:**

In machine learning, you want the **best possible** model, not just any model that works. Without a way to measure "goodness," you cannot improve your model systematically.

**Key term - Quantify:**  
To "quantify" means to measure something with a number. The perceptron trick cannot quantify how good a line is. It only tells you "works" or "doesn't work."

---

#### Problem 2: Convergence Problems

**What is convergence?**  
Convergence means the training process successfully finds a good line and stops changing.

**The Issue:**

Because the perceptron trick randomly selects points, it may:
- Get stuck and never find a good line
- Take too many iterations to find a line
- Keep changing forever without settling

**Hypothetical scenarios where convergence fails:**

| Scenario | What happens | Why it's a problem |
|----------|--------------|---------------------|
| You select the same correctly classified points repeatedly | The line never changes because correct points cause no update | Training does nothing |
| You select only a small set of points | The line only learns from those few points and ignores others | The line may not work for all data |
| The line is already good, but you keep selecting correct points | No updates happen | Wasted time |

**How often do these problems occur?**

- **98% of the time**, the perceptron trick gives you a working result
- **1% of the time**, convergence fails (no good line is found)
- **100% of the time**, you cannot tell if your line is the best (Problem 1 always exists)

---

#### Summary Table: Perceptron Trick vs. What We Need

| Feature | Perceptron Trick | What We Need |
|---------|------------------|---------------|
| Tells if a line works | ✅ Yes (by checking misclassifications) | ✅ Yes |
| Tells how good a line is (quantifies performance) | ❌ No | ✅ Yes |
| Guarantees convergence | ❌ No (may fail 1% of the time) | ✅ Yes |
| Helps find the best possible line | ❌ No | ✅ Yes |

---

### 3. Key Points

- **Problem 1 (Quality measurement)**: The perceptron trick cannot tell you how good your line is compared to other lines. You cannot quantify performance.
- **Problem 2 (Convergence)**: The perceptron trick may fail to find a good line because it randomly selects points. This happens in about 1% of cases.
- The perceptron trick works 98% of the time, but that is not good enough for serious machine learning.
- You need a method that gives you a **number** representing how good your model is.
- You need a method that guarantees you can find the **best possible** model.

## Topic 3: Introduction to Loss Functions

### 1. Introduction

**What is this topic?**  
A loss function is a mathematical way to measure how well or how poorly your machine learning model is performing. It gives you a single number that represents the "error" or "mistake" of your model.

**Why is it important?**  
Loss functions solve both problems of the perceptron trick:
- They **quantify** performance (give you a number to compare models)
- They help you find the **best possible** model by making that number as small as possible

**Where is it used in real life?**  
Every machine learning algorithm uses loss functions:
- Linear regression uses Mean Squared Error (MSE)
- Logistic regression uses Log Loss (Binary Cross-Entropy)
- Neural networks use various loss functions depending on the task
- Even everyday decisions (like "which route is shortest?") use a loss concept

---

### 2. Detailed Explanation

#### What is a Loss Function?

A **loss function** is a mathematical function that:
- Takes the model's parameters (weights w₁, w₂ and bias b) as input
- Produces a **single number** as output
- That number tells you how bad your model is

**Key idea:**  
- **Small loss number** = Good model (few mistakes, small errors)
- **Large loss number** = Bad model (many mistakes, large errors)

#### How Loss Functions Solve the Perceptron Trick Problems

| Problem | How Loss Function Solves It |
|---------|----------------------------|
| Cannot tell which line is better | Calculate loss for each line → lower loss = better line |
| Convergence problems | Use optimization to systematically reduce loss, not random selection |
| Cannot quantify performance | Loss gives a number you can read and compare |

#### The Loss Function Workflow

```
Line (weights w₁, w₂, bias b) → Loss Function → Number (loss value)
                                                      ↓
                                            Lower number = Better line
```

#### Simple Example of a Loss Function

The transcript gives a very simple example:

**Loss Function = Number of misclassified points**

- Draw a line
- Count how many points are on the wrong side
- That count is your loss

**Example calculation:**

| Line | Misclassified points | Loss value |
|------|---------------------|------------|
| Blue line | 7 mistakes | 7 |
| New line | 5 mistakes | 5 |

Since 5 is less than 7, the new line is better.

#### Why This Simple Loss Function is Not Perfect

The simple loss function (counting mistakes) treats all mistakes equally:

- A point far from the line (big mistake) → counted as 1
- A point very close to the line (small mistake) → also counted as 1

**Problem:** A point that is far away should be considered a "worse" mistake than a point that is almost correct.

**Better approach:** Measure the **distance** of misclassified points from the line, not just count them.

#### The Ideal Loss Function Should:

1. Give zero contribution for correctly classified points
2. Give larger numbers for points that are far from the line (bigger mistakes)
3. Give smaller numbers for points that are close to the line (smaller mistakes)

---

### 3. Key Points

- A **loss function** produces a number that quantifies model performance
- **Lower loss** = better model
- Loss functions solve the perceptron trick's problems by providing measurement and direction
- Counting misclassified points is a loss function, but it treats all mistakes equally
- Better loss functions consider the **distance** or **magnitude** of mistakes

---

### 4. General Syntax/Structure of a Loss Function

A loss function generally has this form:

```
Loss = (1/n) × Σ (something that measures error for each point)
```

Where:
- **n** = number of data points (rows in your dataset)
- **Σ** (sigma) = sum over all points (add up for each point)
- **Error measurement** = how wrong the model is for that point

For the simple count-loss:
```
Loss = (1/n) × Σ (1 if point is misclassified, else 0)
```

## Topic 4: Perceptron Loss Function - Hinge Loss

### 1. Introduction

**What is this topic?**  
Hinge loss is the specific loss function used for training perceptrons. It is a mathematical formula that quantifies how bad a perceptron's classification is by considering the distance of misclassified points from the decision boundary.

**Why is it important?**  
This is the actual loss function implemented in machine learning libraries (like scikit-learn's SGDClassifier) for perceptrons. Understanding hinge loss allows you to:
- Train perceptrons properly (not just using the trick)
- Know exactly what the perceptron is trying to minimize
- Build a foundation for understanding other loss functions

**Where is it used in real life?**  
- Support Vector Machines (SVMs) use a variant of hinge loss
- Perceptron implementations in popular ML libraries
- Binary classification problems where you want maximum margin separation

---

### 2. Detailed Explanation

#### The Mathematical Formula for Hinge Loss

According to the transcript, the perceptron loss function (hinge loss) is:

```
Loss = (1/n) × Σ max(0, -yᵢ × zᵢ)
```

Where:
- **n** = number of data points (rows in your dataset)
- **Σ** = sum over all points i = 1 to n
- **max(0, something)** = take the larger of 0 or that something (if negative, use 0)
- **yᵢ** = actual class label for point i (+1 or -1)
- **zᵢ** = w₁x₁ + w₂x₂ + b (the value before the step function)

#### Breaking Down the Formula Step by Step

**Step 1: Calculate zᵢ (the linear output)**
```
zᵢ = (w₁ × x₁) + (w₂ × x₂) + b
```
This is the same calculation you do before applying the step function.

**Step 2: Multiply by the true label yᵢ**
```
yᵢ × zᵢ
```

**Step 3: Make it negative and take max with 0**
```
max(0, -yᵢ × zᵢ)
```

**Step 4: Sum over all points and divide by n**
```
Loss = (1/n) × Σ max(0, -yᵢ × zᵢ)
```

#### What Does This Formula Actually Mean?

The transcript explains that hinge loss works differently depending on whether a point is correctly classified or misclassified:

| Situation | yᵢ × zᵢ value | -yᵢ × zᵢ value | max(0, value) | Contribution to loss |
|-----------|---------------|----------------|---------------|---------------------|
| Correctly classified | Positive | Negative | 0 | **Zero** |
| Misclassified | Negative | Positive | Positive number | **Positive (error)** |

**Important insight:** For correctly classified points, hinge loss contributes **nothing** (zero) to the total loss. For misclassified points, it contributes a positive number.

#### Why Use yᵢ × zᵢ Instead of Distance?

The transcript explains that instead of calculating actual perpendicular distance (which is complex), we use `yᵢ × zᵢ` because:

- It is **proportional to distance** (larger distance → larger value)
- It is much **simpler to compute** (just multiplication, no square roots)
- It is a **dot product** operation, which computers handle efficiently

**Comparison:**

| Method | Calculation | Complexity |
|--------|-------------|------------|
| Actual distance | √[(x₂-x₁)² + (y₂-y₁)²] | Complex, slow |
| yᵢ × zᵢ | w₁x₁ + w₂x₂ + b | Simple, fast |

#### Worked Example

Assume we have a line: `z = 2x₁ + 3x₂ + 4`

**Point A (correctly classified green point):**
- Actual label y = +1 (green)
- Coordinates: x₁ = 2, x₂ = 6
- z = (2×2) + (3×6) + 4 = 4 + 18 + 4 = 26 (positive)
- y × z = (+1) × 26 = 26 (positive)
- -y × z = -26 (negative)
- max(0, -26) = 0 → **No contribution to loss**

**Point B (misclassified red point):**
- Actual label y = -1 (red)
- Coordinates: x₁ = -2, x₂ = -2
- z = (2×-2) + (3×-2) + 4 = -4 + (-6) + 4 = -6 (negative)
- y × z = (-1) × (-6) = 6 (positive)
- -y × z = -6 (negative)
- Wait, this gives zero? Let me re-check carefully.

The transcript actually shows a different case. Let me follow the transcript's example:

For a misclassified point, if y is positive but z is negative:
- y = +1 (green point, should be positive side)
- z = negative (actually on negative side - misclassified)
- y × z = (+1) × (negative) = negative
- -y × z = - (negative) = positive
- max(0, positive) = positive → **Contributes to loss**

For a misclassified point, if y is negative but z is positive:
- y = -1 (red point, should be negative side)
- z = positive (actually on positive side - misclassified)
- y × z = (-1) × (positive) = negative
- -y × z = - (negative) = positive
- max(0, positive) = positive → **Contributes to loss**

**Conclusion:** Any time a point is on the wrong side of the line (misclassified), `y × z` is negative, making `-y × z` positive, so it contributes to the loss.

---

### 3. Key Points

- **Hinge loss formula:** `Loss = (1/n) × Σ max(0, -yᵢ × zᵢ)`
- **zᵢ** = w₁x₁ + w₂x₂ + b (linear output before step function)
- **yᵢ** = actual class label (+1 or -1)
- **Correctly classified points** contribute **0** to loss
- **Misclassified points** contribute a **positive number** proportional to distance
- Using `y × z` is simpler and faster than calculating actual distance
- This is the actual loss function used in library implementations of perceptrons

---

### 4. Syntax/Structure of the Formula

```
       1    n
Loss = — × Σ max(0, -yᵢ × (w₁x₁ᵢ + w₂x₂ᵢ + b))
       n   i=1
```

**Breaking it down:**

| Component | Meaning |
|-----------|---------|
| `1/n` | Average over all points (divide by number of points) |
| `Σ` | Sum symbol - add up the value for each point |
| `max(0, value)` | If value is negative, use 0; if positive, use the value |
| `-yᵢ × (w₁x₁ + w₂x₂ + b)` | Negative of (true label × linear output) |
| `yᵢ` | True label (+1 or -1) |
| `w₁, w₂, b` | Model parameters (what we are trying to find) |
| `x₁, x₂` | Input features (from your data, constant) |

## Topic 5: Geometric Intuition of Hinge Loss (How the Loss Function Works Visually)

### 1. Introduction

**What is this topic?**  
This topic explains the geometric meaning of hinge loss—what the loss function is actually doing visually when you look at points and a line. It shows how each point's position relative to the line determines its contribution to the total loss.

**Why is it important?**  
Understanding the geometry helps you:
- Visualize what the loss function is measuring
- Understand why correctly classified points don't affect the loss
- See why misclassified points contribute based on how far they are from the line
- Build intuition for why minimizing loss works

**Where is it used in real life?**  
- Debugging machine learning models (visualizing why loss is high)
- Explaining model decisions to non-technical stakeholders
- Designing new loss functions for specific problems

---

### 2. Detailed Explanation

#### The Setup

Assume we have a dataset with:
- **Two input columns** (x₁ and x₂, like CGPA and IQ)
- **One output column** (y = +1 for "placed", -1 for "not placed")
- Data points plotted on a 2D graph

#### The Four Possible Types of Points Relative to a Line

The transcript explains that any point can fall into one of four categories:

| Type | Actual Label (y) | Side of Line (z sign) | Classification | Loss Contribution |
|------|-----------------|----------------------|----------------|-------------------|
| 1 | +1 (positive class) | Positive side (z > 0) | Correct | **0** |
| 2 | -1 (negative class) | Negative side (z < 0) | Correct | **0** |
| 3 | +1 (positive class) | Negative side (z < 0) | Misclassified | **Positive** |
| 4 | -1 (negative class) | Positive side (z > 0) | Misclassified | **Positive** |

#### Visual Diagram

```
                    Positive Side (z > 0)
                    Output would be +1
                         |
                         |  Type 1: y=+1, z>0 ✓ CORRECT
                         |  (contributes 0 to loss)
                         |
    ---------------------+---------------------  Decision Boundary (z=0)
                         |        (the line)
                         |
                         |  Type 4: y=-1, z>0 ✗ MISCLASSIFIED
                    Negative Side (z < 0)   (contributes positive to loss)
                    Output would be -1
```

#### Step-by-Step Geometric Explanation

**For a correctly classified point:**

Take a green point (y = +1) that should be placed. If it falls on the positive side of the line (z > 0):

- y = +1, z = positive
- y × z = positive × positive = positive
- -y × z = negative
- max(0, negative) = 0

The transcript states: *"If your line has classified correctly, then the contribution of that point to the loss function will be zero."*

**For a correctly classified negative point:**

Take a red point (y = -1) that should NOT be placed. If it falls on the negative side of the line (z < 0):

- y = -1, z = negative
- y × z = negative × negative = positive
- -y × z = negative
- max(0, negative) = 0

Again, contribution is zero.

**For a misclassified point (green on negative side):**

Take a green point (y = +1) that falls on the negative side (z < 0):

- y = +1, z = negative
- y × z = positive × negative = negative
- -y × z = positive (because negative of negative is positive)
- max(0, positive) = that positive number

**What is that positive number?**  
It equals `-y × z` = `-(+1 × negative)` = `-(negative)` = positive = `|z|`

But more importantly, the transcript explains that when you plug the coordinates of this misclassified point into the line equation `z = w₁x₁ + w₂x₂ + b`, the magnitude you get is proportional to how far the point is from the line.

**For a misclassified point (red on positive side):**

Take a red point (y = -1) that falls on the positive side (z > 0):

- y = -1, z = positive
- y × z = negative × positive = negative
- -y × z = positive
- max(0, positive) = that positive number

#### The Key Insight

The transcript emphasizes: *"For the points that if the point is classified then that point has zero contribution in calculating the loss function, but if the point is misclassified then we are taking -y × z."*

**This means:** The loss function only "cares about" (assigns non-zero value to) points that are on the wrong side of the line. And for those points, the loss is larger when the point is farther from the line.

## Topic 6: Optimization - Finding the Best Weights and Bias (Minimizing Loss with Gradient Descent)

### 1. Introduction

**What is this topic?**  
Optimization is the process of finding the values of weights (w₁, w₂) and bias (b) that make the loss function as small as possible. The specific optimization algorithm discussed in the transcript is called **gradient descent**.

**Why is it important?**  
Simply having a loss function is not enough. You need a systematic way to find the best parameters. Gradient descent is the most common optimization algorithm in machine learning. It tells you how to update your weights and bias step by step to reduce the loss.

**Where is it used in real life?**  
- Training all types of neural networks
- Training linear and logistic regression models
- Any machine learning problem where you need to minimize a function
- Even non-machine learning problems like finding the minimum of any mathematical function

---

### 2. Detailed Explanation

#### The Goal of Optimization

Our job is to find values of w₁, w₂, and b such that:

```
Loss(w₁, w₂, b) is MINIMUM
```

Where Loss = (1/n) × Σ max(0, -yᵢ × (w₁x₁ + w₂x₂ + b))

In other words: We want the line (determined by w₁, w₂, b) that has the smallest possible loss.

#### Visualizing the Optimization Problem

The transcript explains it with a simple analogy:

Imagine you have a function where the loss (L) changes as you change only one weight (w₁). The graph might look like a curve:

```
Loss
  ↑
  |    \      /
  |     \    /
  |      \  /
  |       \/
  +----------------→ w₁
```

The transcript states: *"Your job is to find the value of w for which L is the minimum"* — the lowest point on this curve.

#### What is Gradient Descent?

**Gradient descent** is an optimization algorithm that:
1. Starts with random values for w₁, w₂, and b
2. Repeatedly updates them to reduce the loss
3. Uses something called the **derivative** (or gradient) to know which direction to move

**Analogy from the transcript:**  
Think of standing on a hill in fog. You cannot see the bottom, but you can feel the slope beneath your feet. You take a step in the steepest downward direction. Repeat until you reach the bottom. Gradient descent does the same thing mathematically.

#### The Update Rules

The transcript provides three update rules (one for each parameter):

**For w₁:**
```
w₁(new) = w₁(old) - learning_rate × (∂Loss / ∂w₁)
```

**For w₂:**
```
w₂(new) = w₂(old) - learning_rate × (∂Loss / ∂w₂)
```

**For b:**
```
b(new) = b(old) - learning_rate × (∂Loss / ∂b)
```

Where:
- **∂Loss/∂w₁** (partial derivative) = how much the loss changes when you change w₁ slightly
- **learning_rate** = step size (how big of a step to take)

#### What is Learning Rate?

The transcript mentions a learning rate of **0.1** as an example.

| Learning Rate | Effect |
|---------------|--------|
| Too small | Takes many steps, slow training |
| Too large | Might jump over the minimum, never converge |
| Just right | Finds minimum efficiently |

**Example from transcript:** `learning rate = 0.1`

#### The Training Loop (Algorithm)

The transcript describes a loop that runs many times (e.g., 1000 iterations):

```
Initialize w₁, w₂, b randomly
Set learning_rate = 0.1

For iteration = 1 to 1000:
    Calculate the partial derivatives
    Update w₁ = w₁ - learning_rate × derivative_w1
    Update w₂ = w₂ - learning_rate × derivative_w2
    Update b = b - learning_rate × derivative_b

Return final w₁, w₂, b
```

#### Calculating the Partial Derivatives for Hinge Loss

The transcript shows how to derive the derivatives for the perceptron loss function.

**For weight w₁:**
```
∂Loss/∂w₁ = -yᵢ × x₁  (if point is misclassified)
∂Loss/∂w₁ = 0         (if point is correctly classified)
```

**For weight w₂:**
```
∂Loss/∂w₂ = -yᵢ × x₂  (if point is misclassified)
∂Loss/∂w₂ = 0         (if point is correctly classified)
```

**For bias b:**
```
∂Loss/∂b = -yᵢ        (if point is misclassified)
∂Loss/∂b = 0          (if point is correctly classified)
```

**Important note from the transcript:** The update rule actually has a **plus** sign in the implementation (they corrected themselves: "The update rule for the gradient descent has a plus here"). This depends on how the loss function is written. The key is that you move in the opposite direction of the gradient.

#### Simplified Update Rules (For Perceptron Hinge Loss)

For a **misclassified point** (y × z < 0):

```
w₁ = w₁ + learning_rate × y × x₁
w₂ = w₂ + learning_rate × y × x₂
b  = b  + learning_rate × y
```

For a **correctly classified point**:
- No update (all changes are zero)

This matches the perceptron trick but with an important difference: the magnitude of the update depends on the values, not just a fixed amount.

---

### 3. Key Points

- **Goal**: Find w₁, w₂, b that minimize the loss function
- **Algorithm**: Gradient descent
- **Update rule**: `parameter = parameter - learning_rate × derivative`
- **Partial derivatives**: Tell you how loss changes when you change each parameter
- **Learning rate**: Controls step size (example: 0.1)
- **Training loop**: Repeat updates many times (example: 1000 iterations)
- For correctly classified points: derivatives = 0 (no update)
- For misclassified points: derivatives are non-zero (update occurs)

---

### 4. Algorithm Structure

```
PERCEPTRON TRAINING WITH GRADIENT DESCENT

INPUT:
    X = training data (n rows, 2 columns)
    y = true labels (+1 or -1)
    learning_rate = 0.1
    epochs = 1000

OUTPUT:
    w₁, w₂, b (trained parameters)

PROCEDURE:
    1. Initialize w₁ = random, w₂ = random, b = random
    2. For epoch = 1 to epochs:
    3.     For each point i = 1 to n:
    4.         Calculate z = w₁×x₁ + w₂×x₂ + b
    5.         If y_i × z < 0:    # point is misclassified
    6.             w₁ = w₁ + learning_rate × y_i × x₁
    7.             w₂ = w₂ + learning_rate × y_i × x₂
    8.             b  = b  + learning_rate × y_i
    9.         # If correctly classified, do nothing
    10. Return w₁, w₂, b
```

## Topic 8: Flexibility of the Perceptron Model (Different Activation and Loss Functions)

### 1. Introduction

**Why is it important?**

**Where is it used in real life?**  
- **Logistic regression** (binary classification with probabilities)
- **Softmax regression** (multi-class classification)

- **Neural networks** (which extend this flexibility with multiple layers)

#### The Core Perceptron Architecture

The transcript describes the basic perceptron architecture:

The transcript states: *"The perceptron is a mathematical model. This model of perceptron gets its input, summation happens, then it goes to activation function, and then you implement a loss function here for training."*

What can you change?
| Component | What you can change it to |
|-----------|---------------------------|

### 3. Detailed Breakdown of Each Case

  - If z < 0 → output = -1

- **Output**: Either +1 or -1 (hard classification)
- **Use case**: Simple binary classification where you just need the class, not probability

#### Case 2: Logistic Regression (Binary Classification with Probabilities)

The transcript explains: *"If I want the probabilities of the output being one and minus one, I can replace this step function with the sigmoid function."*

- **Activation function**: Sigmoid (also called logistic function)
  - Formula: `σ(z) = 1 / (1 + e^(-z))`
  - Output is always between 0 and 1
  - Can be interpreted as probability
- **Loss function**: Binary Cross-Entropy (also called Log Loss)
  - Formula: `Loss = -[y × log(ŷ) + (1-y) × log(1-ŷ)]`
- **Output**: Probability between 0 and 1
- **The transcript states**: *"You have seen this combination before in machine learning algorithms and the name of that machine learning algorithm is logistic regression."*

#### Case 3: Softmax Regression (Multi-Class Classification)

- **Activation function**: Softmax
  - Formula: `softmax(zⱼ) = e^(zⱼ) / Σ e^(zₖ)` (sum over all classes k)
  - Converts multiple z values into probabilities that sum to 1
- **Loss function**: Categorical Cross-Entropy
  - Formula: `Loss = - Σ y_c × log(ŷ_c)` (sum over all classes c)
- **Output**: Probabilities for each class (sum = 1)
- **Use case**: When you have more than 2 classes (e.g., classify image as cat, dog, or bird)

#### Case 4: Linear Regression (Predicting Numbers)

- **Activation function**: Linear (or no activation function)
  - The transcript says: *"It will take z and give the same z as output. So basically, there is no activation function."*
  - Output = z (no transformation)
- **Loss function**: Mean Squared Error (MSE)
  - Formula: `Loss = (1/n) × Σ (y - ŷ)²`
- **Output**: Any real number (can be 100, 4.5, -10, etc.)
- **Use case**: Predicting continuous values (house prices, temperature, etc.)

### 5. Key Points

- The perceptron is a **flexible architecture**, not a single algorithm
- You can swap **activation functions** and **loss functions** independently
- Different combinations produce different well-known algorithms
- **Step + Hinge** = Classic Perceptron
- **Sigmoid + Binary Cross-Entropy** = Logistic Regression
- **Softmax + Categorical Cross-Entropy** = Softmax Regression
- **Linear + MSE** = Linear Regression
- All use the same **gradient descent** training mechanism

### 6. Formula Structure

**Sigmoid Function (Logistic Regression):**
```
σ(z) = 1 / (1 + e^(-z))
```
- Output range: 0 to 1
- Interpret as probability of positive class

**Binary Cross-Entropy Loss:**
```
Loss = -[y × log(ŷ) + (1-y) × log(1-ŷ)]
```
- y = true label (0 or 1)
- ŷ = predicted probability (from sigmoid)

**Softmax Function (for K classes):**
```
softmax(z_j) = e^(z_j) / Σ(k=1 to K) e^(z_k)
```
- Output for each class j
- All outputs sum to 1

**Categorical Cross-Entropy Loss:**
```
Loss = - Σ(j=1 to K) y_j × log(ŷ_j)
```
- y_j = 1 if class j is correct, else 0 (one-hot encoding)
- ŷ_j = predicted probability for class j

**Mean Squared Error:**
```
Loss = (1/n) × Σ (y_i - ŷ_i)²
```
- y_i = true value
- ŷ_i = predicted value

### 7. Common Mistakes

| Mistake | Why it happens | How to avoid |
|---------|----------------|---------------|
| Using sigmoid with hinge loss | Not understanding pairings | Sigmoid pairs with cross-entropy; step pairs with hinge |
| Using softmax for binary classification | Overcomplicating | For binary, sigmoid is simpler and equivalent |
| Forgetting to convert labels | Different algorithms expect different label formats | Perceptron: -1/+1; Logistic: 0/1 |
| Thinking perceptron = logistic regression always | Not understanding the flexibility | They are the same ONLY when activation=sigmoid and loss=cross-entropy |

### 8. Interview/Exam Questions

**Q1: How can the same perceptron model become logistic regression?**  
A: By changing the activation function from step to sigmoid and changing the loss function from hinge loss to binary cross-entropy.

**Q2: What activation function and loss function would you use for a multi-class classification problem?**  
A: Softmax activation function with categorical cross-entropy loss.

**Q3: Can a perceptron be used for regression problems? How?**  
A: Yes. Use a linear activation function (or no activation function) and mean squared error (MSE) as the loss function.

**Q4: What is the output range of the sigmoid function? What does it represent?**  
A: The sigmoid function outputs values between 0 and 1, which can be interpreted as probabilities.

**Q5: Why is the perceptron considered a "flexible" model?**  
A: Because its architecture separates the summation, activation, and loss functions. You can swap different activation and loss functions to solve different types of problems (binary classification, multi-class classification, regression) without changing the core training mechanism.
