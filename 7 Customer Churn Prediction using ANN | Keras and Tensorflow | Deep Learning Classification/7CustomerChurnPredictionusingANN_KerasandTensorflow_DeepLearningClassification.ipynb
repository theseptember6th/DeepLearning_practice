## Topic 4: Train-Test Split & Feature Scaling (Standardization)

### 1. Introduction

- **What is this topic?**
    This topic covers two crucial steps before building a neural network:
    1. **Train-Test Split:** Dividing our data into two parts – one for training the model and one for testing it.
    2. **Feature Scaling:** Transforming all numerical values so they are on a similar scale (e.g., between -1 and 1 or 0 and 1).

- **Why is it important?**
    - If you test your model on the same data you trained on, you won't know if it actually learned or just memorized the answers.
    - Neural networks struggle to learn when numbers have very different ranges (e.g., Age: 20-80 vs Salary: 20,000-200,000). Scaling fixes this.
    - Without scaling, the model may take much longer to train or may never learn properly.

- **Where is it used in real life?**
    - Every machine learning project uses train-test split.
    - Feature scaling is especially important in neural networks, image processing, and any algorithm that uses distance or gradient calculations.

---

### Part A: Train-Test Split

**What is the problem?**
- We have 10,000 customer records.
- If we train the model on ALL 10,000 and then test it on the SAME 10,000, the model might just memorize the answers.
- When we give it a NEW customer (not in the training data), it will fail.

**What is the solution?**
- Split the data into two separate groups:

| Group | Percentage | Purpose | Analogy |
| --- | --- | --- | --- |
| **Training Set** | 80% (8,000 customers) | The model learns patterns from this data | Like studying textbooks before an exam |
| **Testing Set** | 20% (2,000 customers) | The model is tested on this unseen data | Like taking the final exam on new questions |

### Part B: Feature Scaling (Standardization)

**What is the problem?**
Look at these different columns in our dataset:

| Column | Typical Range |
| --- | --- |
| Age | 18 to 100 |
| Credit Score | 300 to 850 |
| Balance | 0 to 250,000 |
| Estimated Salary | 10,000 to 200,000 |
| Tenure | 0 to 10 |
| Num of Products | 1 to 4 |

**The problem:** These numbers are on completely different scales. A neural network works by doing math (multiplication and addition). When one number is 100,000 and another is 0.5, the large number "overwhelms" the small number.

**What happens without scaling?**
- The model focuses too much on large numbers (like Balance)
- It ignores small numbers (like Tenure)
- The model takes much longer to learn (converges slowly)
- The model may never find the optimal solution

**What is the solution? Standardization (Z-score scaling)**

Standardization transforms each column so that:
- **Mean (average) becomes 0**
- **Standard deviation becomes 1**

**The formula:**
```
Scaled Value = (Original Value - Mean) / Standard Deviation
```

**Simple example:**
Suppose a column has values: [20, 30, 40]
- Mean = 30
- Standard Deviation = 10

Apply formula:
- (20 - 30) / 10 = -1.0
- (30 - 30) / 10 = 0
- (40 - 30) / 10 = 1.0

After scaling: [-1.0, 0, 1.0]

**Code from transcript:**
```python
from sklearn.preprocessing import StandardScaler

# Create the scaler object
scaler = StandardScaler()

# Fit the scaler to training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test data using the SAME scaler
X_test_scaled = scaler.transform(X_test)
```

**Why do we use `fit_transform` on training but only `transform` on testing?**
- `fit_transform` learns the mean and standard deviation from training data AND applies scaling
- `transform` only applies scaling using the mean and standard deviation learned from training
- **Important:** We must NOT "fit" on test data because in real life, we don't know the mean of future customers

**Result after scaling:**
- All columns now have similar ranges (approximately -2 to +2)
- No single column dominates the others
- The neural network learns much faster and better

---

### 3. Key Points

**Train-Test Split:**
- Training set: 80% of data (8,000 customers)
- Testing set: 20% of data (2,000 customers)

- `test_size=0.2` means 20% for testing

**Feature Scaling:**
- Neural networks need scaled data to learn properly
- Standardization makes mean = 0, standard deviation = 1
- Formula: (value - mean) / standard deviation
- Fit scaler ONLY on training data, then transform both training and test data

| Testing on training data | If you test on the same data used for training, you get over-optimistic results that don't reflect real-world performance | Always split data BEFORE training |
| Fitting scaler on test data | Using `fit_transform` on test data means test data's mean influences scaling, which is cheating | Use `fit_transform` ONLY on training, then `transform` on test |
| Forgetting to scale | Neural networks work poorly with unscaled data | Always scale before building neural network |
| Scaling target (y) | In classification, the target values (0 and 1) do not need scaling | Only scale features (X), not target (y) |
| Not keeping the split consistent | Each time you run the code, you get a different random split | Use `random_state=42` in `train_test_split` for reproducible results |

### 8. Interview/Exam Questions

**Q1: Why do we split data into training and testing sets?**

**A1:** We split data to evaluate how well our model performs on unseen data. If we test on the same data used for training, we cannot tell if the model actually learned general patterns or simply memorized the answers (overfitting).

**Q2: What is the typical ratio for train-test split in machine learning?**

**A2:** Common ratios are 80-20 (80% training, 20% testing) or 70-30. The transcript uses 80-20.

**Q3: Why do neural networks require feature scaling?**

**A3:** Neural networks use gradient descent to learn. When features have different scales, the gradient descent path becomes very long and winding, making training slow. Scaling creates a smoother path, allowing faster and more reliable training.

**Q4: What is the difference between `fit_transform()` and `transform()`?**

**A4:** 
- `fit_transform()`: Calculates parameters (mean, standard deviation) from the data AND applies the transformation
- `transform()`: Only applies transformation using existing parameters (does NOT recalculate)

**Q5: Why must we fit the scaler only on training data, not on test data?**

**A5:** Because in real-world deployment, new data comes without any pre-computed statistics. We want to simulate that by using only training statistics to scale test data. Scaling test data with its own statistics would be "cheating" because it gives the model information from the test set.

- Purpose: Bring all features to same scale

- Result: Mean = 0, Standard Deviation = 1

## Topic 6: Compiling the ANN (Loss Function, Optimizer, Metrics)

### 1. Introduction

- **What is this topic?**
    Compiling is the step where we configure how the neural network will learn. We specify three things:
    1. **Optimizer:** The algorithm that updates the weights to reduce errors
    2. **Loss Function:** A way to measure how wrong the model's predictions are
    3. **Metrics:** What numbers we want to track during training (like accuracy)

- **Why is it important?**
    - Without compiling, the model doesn't know how to learn.
    - Choosing the right loss function and optimizer determines whether your model learns successfully or fails.
    - This is where the "learning" part of machine learning is configured.

- **Where is it used in real life?**
    - Every neural network in every application (facial recognition, language translation, stock prediction) must be compiled before training.
    - Different problems require different loss functions.

---

### 2. Detailed Explanation

Let's understand each component step by step.

---

**Step 1: What is a Loss Function?**

- **Loss** = a number that tells us how wrong the model's predictions are.
- **Low loss** = predictions are close to actual values (good)
- **High loss** = predictions are far from actual values (bad)
- The goal of training is to **minimize the loss**.

**Analogy:** 
Imagine you are throwing darts at a bullseye.
- Each throw = a prediction
- The distance from the bullseye = the loss
- Your goal is to throw darts as close to the bullseye as possible (minimize the distance/loss)

**What loss function do we use for binary classification?**

From the transcript: *"binary cross entropy which you can also call log loss"*

| Loss Function | Full Name | When to use |
| --- | --- | --- |
| **binary_crossentropy** | Binary Cross Entropy | Binary classification (2 outcomes: yes/no, leave/stay) |
| categorical_crossentropy | Categorical Cross Entropy | Multi-class classification (3+ categories) |
| mean_squared_error | MSE | Regression problems (predicting numbers) |

**How does binary cross entropy work?**
- It compares the predicted probability (0 to 1) with the actual answer (0 or 1).
- If the prediction is correct and confident → low loss
- If the prediction is wrong or uncertain → high loss

**Example:**
| Actual (y_true) | Predicted (y_pred) | Loss |
| --- | --- | --- |
| 1 (left) | 0.90 | Very low (correct, confident) |
| 1 (left) | 0.51 | Low (correct, but not confident) |
| 1 (left) | 0.30 | High (WRONG) |
| 0 (stayed) | 0.10 | Very low (correct, confident) |
| 0 (stayed) | 0.80 | High (WRONG) |

---

**Step 2: What is an Optimizer?**

- The **optimizer** is the algorithm that updates the model's weights to reduce the loss.
- It figures out in which direction to change each weight to make the loss smaller.

**Analogy:**
Imagine you are blindfolded on a mountain and want to reach the bottom (minimum loss).
- You feel the ground with your feet to sense which direction goes downhill.
- You take a small step in that direction.
- You repeat until you reach the bottom.
- The optimizer is like your "downhill sensing" and "stepping" strategy.

**What optimizer do we use?**

From the transcript: *"Adam generally performs better"*

| Optimizer | Full Name | Characteristics |
| --- | --- | --- |
| **Adam** | Adaptive Moment Estimation | Most popular; adapts learning rate automatically; works well for most problems |
| SGD | Stochastic Gradient Descent | Basic optimizer; slower; requires tuning learning rate |
| RMSprop | Root Mean Square Propagation | Good for recurrent neural networks |

**Why Adam?**
- It is adaptive: it uses different learning rates for different parameters.
- It combines the benefits of two other optimizers (AdaGrad and RMSprop).
- It works well out-of-the-box with minimal tuning.

---

**Step 3: What are Metrics?**

- **Metrics** are additional numbers we track to see how well the model is performing.
- Unlike loss (which is used for training), metrics are just for us to monitor.

**Example:**

### 8. Interview/Exam Questions

**Q1: What is the difference between loss function and a metric?**

**A1:** 
- **Loss function:** Used by the optimizer to update weights during training. Must be differentiable (mathematically smooth).
- **Metric:** Used for humans to monitor performance. Does not affect training. Can be non-differentiable (like accuracy).

**Q2: Why do we use binary cross-entropy for this problem instead of categorical cross-entropy?**

**A2:** Binary cross-entropy is designed for binary classification problems where there are only two possible outcomes (leave or stay). Categorical cross-entropy is for multi-class problems with three or more categories.

**Q3: What does the Adam optimizer do in simple terms?**

**Q5: What is the relationship between loss and accuracy?**

**A5:** Generally, when loss decreases, accuracy increases. However, it's possible to have low loss but moderate accuracy, or high loss but decent accuracy. Both should be monitored together.

## Topic 7: Training the ANN (Model Fitting)

### 1. Introduction

- **What is this topic?**
    Training (also called "fitting") is the process where the neural network learns from the data. We feed the model our training data (X_train_scaled and y_train), and the model repeatedly adjusts its internal weights to minimize the loss. Each complete pass through all the training data is called an **epoch**.

- **Why is it important?**
    - This is where the actual "learning" happens.
    - Without training, the model's predictions are random and useless.
    - Understanding epochs and batch sizes helps you train models efficiently without wasting time.

- **Where is it used in real life?**
    - Training is the most time-consuming and computationally expensive part of any deep learning project.
    - Large models (like ChatGPT) are trained for weeks or months on thousands of computers.

---

### 2. Detailed Explanation

Let's understand the training process step by step.

---

**Step 1: What happens during training?**

During training, the model does this repeatedly:

1. **Forward pass:** Data flows through the network to make a prediction.
2. **Calculate loss:** Compare prediction with actual answer (y_train).
3. **Backward pass (backpropagation):** Calculate how much each weight contributed to the error.
4. **Update weights:** The optimizer adjusts weights slightly to reduce future errors.

This happens for every batch of data, for each epoch.

---

**Step 2: What is an Epoch?**

An **epoch** = one complete pass through ALL training data.

**Analogy:**
Imagine you are studying a textbook of 100 pages for an exam.
- One **epoch** = reading all 100 pages once (from page 1 to page 100)
- If you read the book 10 times = 10 epochs

**From the transcript:** *"10 epochs... 10 times"*

| Epochs | What it means |
| --- | --- |
| 1 | Model sees each training example once |
| 10 | Model sees each training example 10 times |
| 50 | Model sees each training example 50 times |

**Important:** More epochs = more learning time, but too many can cause overfitting (memorizing instead of learning patterns).

---

**Step 3: What is Batch Size?**

- The model cannot process all 8,000 training examples at once (memory limits).
- So data is split into smaller groups called **batches**.
- **Batch size** = number of examples processed before updating weights.

**From the transcript:** *The transcript mentions "10 times" but doesn't explicitly specify batch size. Typically, a default like 32 is used.*

**Example with batch_size = 32:**
- Training data: 8,000 customers

**Batch size comparison:**

| Medium (e.g., 64-128) | Balanced | Good | Medium |

## Topic 9: Evaluating Model Performance (Confusion Matrix & Accuracy)

### 1. Introduction

- **What is this topic?**
    After making predictions, we need to evaluate how well our model performed. We compare the model's predictions (y_pred) with the actual answers (y_test) using metrics like **accuracy** and the **confusion matrix**. These tell us how many predictions were correct and, more importantly, what kinds of mistakes the model made.

- **Why is it important?**
    - Accuracy alone can be misleading, especially with imbalanced data (like our 80% stay, 20% leave).
    - A confusion matrix shows exactly where the model succeeded and failed.
    - Understanding these metrics helps you decide if your model is good enough for real-world use.

- **Where is it used in real life?**
    - Banks need to know: "Of the customers we flagged as high-risk, how many actually left?"

### 2. Detailed Explanation

Let's understand evaluation step by step.

---

**Step 1: What is a Confusion Matrix?**

A confusion matrix is a table that shows the counts of correct and incorrect predictions broken down by each class.

**For binary classification (0 = Stay, 1 = Leave):**

| | **Predicted: STAY (0)** | **Predicted: LEAVE (1)** |
| --- | --- | --- |
| **Actual: STAY (0)** | True Negatives (TN) | False Positives (FP) |
| **Actual: LEAVE (1)** | False Negatives (FN) | True Positives (TP) |

**Let's understand each term:**

| Term | Meaning | In our project |
| --- | --- | --- |
| **True Positive (TP)** | Model predicted LEAVE, and customer actually LEFT | ✅ Correct prediction |
| **True Negative (TN)** | Model predicted STAY, and customer actually STAYED | ✅ Correct prediction |
| **False Positive (FP)** | Model predicted LEAVE, but customer STAYED | ❌ False alarm (Type 1 Error) |
| **False Negative (FN)** | Model predicted STAY, but customer LEFT | ❌ Missed detection (Type 2 Error) |

**Simple memory trick:**
- "True" = prediction was correct
- "False" = prediction was wrong
- "Positive" = predicted LEAVE (1)
- "Negative" = predicted STAY (0)

```python
from sklearn.metrics import confusion_matrix, accuracy_score

# Assume y_test is actual values, y_pred is predictions
# (y_pred would have been created earlier using threshold)

# Calculate confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Extract individual values
tn, fp, fn, tp = cm.ravel()
print(f"\nTrue Negatives (TN): {tn}")
print(f"False Positives (FP): {fp}")
print(f"False Negatives (FN): {fn}")
print(f"True Positives (TP): {tp}")

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nAccuracy: {accuracy * 100:.2f}%")

# Calculate additional metrics (optional)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"\nPrecision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1-Score: {f1_score * 100:.2f}%")
```

## Topic 11: Using Validation Data & Plotting Training History

### 1. Introduction

- **What is this topic?**
    During training, we can set aside a portion of the training data as a **validation set**. The model does not learn from this validation data – it is only used to monitor performance during training. We can then plot **learning curves** (graphs of loss and accuracy over epochs) to see how well the model is learning and detect overfitting.

- **Why is it important?**
    - Without validation, you cannot tell if your model is overfitting.
    - Learning curves give you a clear visual of your model's progress.
    - This helps you decide when to stop training (early stopping) and which model architecture works best.

- **Where is it used in real life?**
    - Every deep learning researcher uses validation sets and learning curves.
    - Companies use these plots to compare different model versions.
    - Automated systems use validation loss to trigger early stopping.

---

### 2. Detailed Explanation

Let's understand validation and plotting step by step.

---

**Step 1: What is a Validation Set?**

We already split data into **training (80%)** and **testing (20%)** . But during training, we can further split the training data:

| Dataset | Percentage | Purpose |
| --- | --- | --- |
| Training | 64% (of total) | Model learns from this data |
| Validation | 16% (of total) | Monitor during training (no learning) |
| Testing | 20% (of total) | Final evaluation after training |

**From the transcript:**
*"I am separating 20% of customers from my 8000 training customers to use as validation"*

**Code from transcript:**
```python
history = model.fit(X_train_scaled, y_train, epochs=100, validation_split=0.2)
```

**What `validation_split=0.2` means:**
- Take the training data (X_train_scaled, y_train)
- Randomly split off 20% of it for validation
- The model trains on the remaining 80% of training data
- After each epoch, the model is evaluated on the validation set

**Important distinction:**
| | Validation Set | Test Set |
| --- | --- | --- |
| Used during training? | Yes (for monitoring) | No (only after training) |
| Model learns from it? | No | No |
| Affects model updates? | No (just for monitoring) | No |
| When to use? | Every epoch | Once at the end |

---

**Step 2: The History Object**

When we add `validation_split`, the `history` object stores both training and validation metrics.

**Code:**
```python
history = model.fit(X_train_scaled, y_train, epochs=100, validation_split=0.2)
```

The transcript shows how to plot graphs to visualize training progress.

**Code from transcript:**

**What overfitting looks like:**

| Epochs | Training Loss | Validation Loss | Interpretation |

```
Loss
  ↑
  |     Training Loss (keeps going down)
  |    ╱
  |   ╱
  |  ╱    Validation Loss (goes down, then UP)
  | ╱    ╱‾‾‾‾‾‾
  |╱    ╱
  └──────────────────→ Epochs
```

This small gap between training and validation loss indicates some overfitting, but not severe.

**Step 6: Why Validation is Important**

Without validation:
- You only see training loss (always decreases)
- You cannot tell if model is overfitting
- You might train too long and make the model worse

With validation:
- You see both training and validation loss
- You can stop training when validation loss stops improving
- This technique is called **early stopping**
