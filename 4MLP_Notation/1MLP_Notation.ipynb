# Topic 1: Trainable Parameters in a Neural Network

## 1. Introduction

**What is this topic?**  
This topic teaches you how to count the **number of trainable parameters** in a neural network just by looking at its architecture (diagram). Trainable parameters are the **weights** and **biases** that the network learns during training.

**Why is it important?**  
- Before you train a neural network, you should know how many parameters will be learned.
- This helps you understand the **complexity** of your model.
- It is the first step before understanding backpropagation (the algorithm that trains the network).

**Where is it used in real life?**  
- Designing neural networks for image recognition, speech recognition, or natural language processing.
- Estimating how much memory and computation power a model will need.
- Comparing different network architectures before choosing one.

---

## 2. Detailed Explanation

### What are Trainable Parameters?

When you train a neural network, an algorithm (called **backpropagation**) goes through the network and **adjusts** certain numbers to make the network smarter. These numbers that get adjusted are called **trainable parameters**.

There are **two types** of trainable parameters:

| Parameter Type | What it does | Symbol in diagrams |
|----------------|--------------|--------------------|
| **Weight** | Controls the strength of the connection between two neurons | Lines connecting neurons |
| **Bias** | An extra number added to each neuron (except input layer) to shift the output | Not always drawn, but exists for each neuron (except input) |

> **Beginner definition:**  
> - **Weight** = How much importance one neuron gives to another neuron's signal.  
> - **Bias** = A "nudge" that helps the neuron fire even when all inputs are zero.

### Counting Weights – Step by Step

**Rule:** Each connection (line) between two neurons represents **one weight**.

**Step 1:** Look at the neural network diagram.  
**Step 2:** Count how many lines go from **Layer 0 (input layer)** to **Layer 1 (first hidden layer)**.  
**Step 3:** Count how many lines go from **Layer 1 to Layer 2** (if present).  
**Step 4:** Count how many lines go from the last hidden layer to the **output layer**.  
**Step 5:** Add all these numbers together.

**Example from transcript (simple network):**

```
Layer 0 (Input)     Layer 1 (Hidden)     Layer 2 (Output)
    ○                 ○                     ○
    ○                 ○                     
    ○                 ○
    ○
```

- Layer 0 has **4 neurons**
- Layer 1 has **3 neurons**
- Layer 2 has **2 neurons**

**Counting weights:**

| Connection | Calculation | Number of Weights |
|------------|-------------|-------------------|
| Layer 0 → Layer 1 | 4 × 3 = 12 | 12 |
| Layer 1 → Layer 2 | 3 × 2 = 6 | 6 |
| **Total weights** | 12 + 6 = **18** | 18 |

### Counting Biases

**Rule:** Every neuron **except those in the input layer** has exactly **one bias**.

**Step 1:** Count all neurons in Layer 1 (first hidden layer).  
**Step 2:** Count all neurons in Layer 2 (second hidden layer, if present).  
**Step 3:** Count all neurons in the output layer.  
**Step 4:** Add these numbers together.

**Example from above:**

| Layer | Number of Neurons | Biases in this layer? |
|-------|------------------|----------------------|
| Layer 0 (Input) | 4 | No biases |
| Layer 1 (Hidden) | 3 | 3 biases |
| Layer 2 (Output) | 2 | 2 biases |
| **Total biases** | 3 + 2 = **5** | 5 |

### Total Trainable Parameters

**Formula:**
```
Total Trainable Parameters = Total Weights + Total Biases
```

**From example:**
```
Total Weights = 18
Total Biases = 5
Total Trainable Parameters = 18 + 5 = 23
```

The transcript mentions a total of **30** in one example (when a different architecture is used). The exact number depends on the diagram.

---

## 3. Key Points

- **Trainable parameters** = weights + biases that the network learns during training.
- **Weights** = one per connection between neurons.
- **Biases** = one per neuron (except input layer).
- To calculate weights: multiply number of neurons in connected layers, then sum across all connections.
- To calculate biases: add number of neurons in all non-input layers.
- Knowing this number helps you understand model complexity before training.

## 7. Common Mistakes

| Mistake | Why it's wrong | How to avoid |
|---------|----------------|--------------|
| Counting biases in input layer | Input layer neurons do not have biases | Remember: only hidden and output layers have biases |
| Forgetting to multiply when counting weights | Each neuron in next layer connects to **all** neurons in previous layer | Always multiply: (neurons in previous layer) × (neurons in current layer) |
| Only counting weights, forgetting biases | Both weights and biases are trainable | Always do: total = weights + biases |
| Adding wrong numbers | Mistaking the diagram's neuron count | Label each layer's neuron count before calculating |

---

## 8. Interview/Exam Questions

**Q1:** What are trainable parameters in a neural network?  
**A1:** They are the weights (connections between neurons) and biases (one per non-input neuron) that are adjusted during training using backpropagation.

**Q2:** A neural network has 3 input neurons, 4 hidden neurons, and 2 output neurons. How many trainable parameters does it have?  
**A2:**  
- Weights: (3×4) + (4×2) = 12 + 8 = 20  
- Biases: 4 (hidden) + 2 (output) = 6  
- Total = 20 + 6 = **26 trainable parameters**

**Q3:** Why don't input layer neurons have biases?  
**A3:** Biases are added to the **weighted sum** before applying an activation function. Input layer just passes raw data forward; it does no computation, so it needs no bias.

```
TRAINABLE PARAMETERS = WEIGHTS + BIASES

WEIGHTS:
- One per connection line between neurons
- Count = Σ (neurons_in_prev_layer × neurons_in_current_layer)

BIASES:
- One per neuron NOT in input layer
- Count = total neurons - input_neurons

EXAMPLE:
Input: 4 nodes
Hidden: 3 nodes
Output: 2 nodes

Weights = (4×3) + (3×2) = 12 + 6 = 18
Biases = 3 + 2 = 5
TOTAL = 23

FORMULA:
Total = Σ(Lₙ₋₁ × Lₙ) + Σ(Lₙ) [for n ≥ 1]
```

# Topic 2: Weight Notation (w_ijk)

## 1. Introduction

**What is this topic?**  
Weight notation is a **standard way of naming weights** in a neural network so that you can identify exactly which connection a weight belongs to. The transcript uses the notation **w_ijk** (also written as w[i][j][k] or w with three subscripts).

**Why is it important?**  
- When you write code to train a neural network, you need to update specific weights using backpropagation.
- Without a clear naming system, you will get confused about which weight is which.
- Understanding this notation makes backpropagation (the training algorithm) much easier to learn.

**Where is it used in real life?**  
- In deep learning frameworks like TensorFlow and PyTorch (they use similar indexing internally).
- When manually implementing backpropagation from scratch.
- When reading research papers that describe neural network architectures.

---

## 2. Detailed Explanation

### The Problem That Notation Solves

Imagine you have a neural network with multiple layers. Each layer has multiple neurons. Between every two neurons from adjacent layers, there is a **weight**.

If you have 3 layers, you might have 12, 18, or even 100+ weights. How do you give each weight a unique name so you know exactly:
- **Which layer** the weight belongs to?
- **Which neuron** the weight is coming from?
- **Which neuron** the weight is going to?

The answer is the **w_ijk** notation.

### Understanding the Three Subscripts

The notation is written as: **w_ijk** or **w[i][j][k]**

Each subscript (small index number) has a specific meaning:

| Subscript | Name | What it tells you |
|-----------|------|-------------------|
| **i** | Layer index | Which layer the weight is **going into** (target layer) |
| **j** | From-neuron index | Which neuron the weight is **coming from** (in previous layer) |
| **k** | To-neuron index | Which neuron the weight is **going to** (in current layer) |

> **Beginner memory trick:** Think of i as "**i**ncoming layer", j as "**j**umping from", k as "**k**oming to".

### Visual Analogy: A Road Network

Think of a neural network as a road system between cities:

- **Layers** = different regions (Region 0, Region 1, Region 2)
- **Neurons** = specific cities within a region
- **Weights** = roads connecting cities
- **w_ijk** = a road that starts at city j in one region and ends at city k in the next region

If you say **w_132**, you are describing:  
"A road coming from region 1 (layer 1), city number 3, going to region 2 (layer 2), city number 2"

### Step-by-Step Example from the Transcript

Look at this simple network:

```
Layer 1               Layer 2
(previous layer)      (current layer)
   
Neuron 1 (j=1)  ---w--->  Neuron 1 (k=1)
Neuron 2 (j=2)  ---w--->  Neuron 2 (k=2)
Neuron 3 (j=3)  ---w--->  
```

**If you want to name the weight that connects from Neuron 2 in Layer 1 to Neuron 1 in Layer 2:**

- i = 2 (because weight is going **into** Layer 2)
- j = 2 (coming **from** Neuron 2 in the previous layer)
- k = 1 (going **to** Neuron 1 in the current layer)

**The name is: w_221** (or w[2][2][1])

### Another Example from the Transcript

The transcript gives this example: **w_112**

Let's decode it:

| Part | Value | Meaning |
|------|-------|---------|
| i = 1 | 1 | Weight is going into Layer 1 |
| j = 1 | 1 | Weight is coming from Neuron 1 of previous layer |
| k = 2 | 2 | Weight is going to Neuron 2 of current layer (Layer 1) |

**Translation:** The weight that connects from Neuron 1 of the layer **before** Layer 1 to Neuron 2 of **Layer 1**.

### Important Caveat from the Transcript

The transcript mentions: *"k means what is your weight — which layer it is penetrating into"*

Some sources reverse the order of subscripts. Always check the definition being used. The transcript defines:
- **i** = target layer number
- **j** = source neuron number (from previous layer)
- **k** = target neuron number (in current layer)

---

## 3. Key Points

- **w_ijk** is a naming system for weights in a neural network.
- **i** = the layer the weight is **entering** (target layer)
- **j** = the neuron number the weight is **coming from** (source neuron)
- **k** = the neuron number the weight is **going to** (target neuron)
- This notation allows you to uniquely identify every weight in the network.
- Knowing this notation is essential before learning backpropagation.

---

## 4. Syntax/Structure

| Component | Syntax | Meaning |
|-----------|--------|---------|
| Weight symbol | **w** | Represents a weight parameter |
| First subscript | **w_i** | i = target layer number (which layer this weight goes into) |
| Second subscript | **w_ij** | j = source neuron number (where weight comes from) |
| Third subscript | **w_ijk** | k = target neuron number (where weight goes to) |

**Complete structure:**
```
w_targetLayer_sourceNeuron_targetNeuron
     ↑           ↑             ↑
     i           j             k
```

**Example:**
```
w_132
```
- Target layer = 1
- Source neuron = 3  
- Target neuron = 2

**In code (pseudocode):**
```python
weight[target_layer][source_neuron][target_neuron] = 0.5
```

## 7. Common Mistakes

| Mistake | Why it's wrong | How to avoid |
|---------|----------------|--------------|
| Confusing j and k | j is **from** (source), k is **to** (target) | Remember: j comes before k in the alphabet — source comes before target |
| Thinking i is the source layer | i is the **target** layer (where weight is going into) | Think "**i**ncoming layer" |
| Using different subscript orders inconsistently | Different tutorials use different conventions | Always check the definition at the start of a lesson |
| Forgetting that i refers to the target layer, not the layer number of the source | The weight exists between two layers; you need a consistent reference point | Remember: A weight "belongs to" the layer it enters |

**Q1:** What does w_243 represent in a neural network?

- i = 2 → weight is entering Layer 2

**Q2:** Why do we need three subscripts to name a weight?

# Topic 3: Bias and Output Notation

## 1. Introduction

**What is this topic?**  
This topic explains how to **name biases** and **outputs** of neurons in a neural network using a clear, consistent notation. Just as weights have notation (w_ijk), biases and outputs also follow specific naming rules.

**Why is it important?**  
- Biases are trainable parameters just like weights — you need to know which bias belongs to which neuron.
- Outputs from one layer become **inputs** to the next layer, so you must track them clearly.
- Without proper notation, you cannot write backpropagation algorithms correctly.

**Where is it used in real life?**  
- Implementing neural networks from scratch in code.
- Debugging why a network is not learning properly.
- Understanding how data flows through a network during forward propagation.

---

## 2. Detailed Explanation

### Part A: Bias Notation

**What is a bias?**  
A bias is an **extra number** added to each neuron (except input layer neurons) before applying the activation function. It allows the neuron to "fire" even when all inputs are zero.

**Bias Notation from the Transcript**

The transcript introduces bias notation using: **b_ij** (or similar pattern with two subscripts)

Let's break down what the transcript tells us:

> *"If we come here, there are 3 and if there are two here then we will think about the total rate... and then its virus"* (the transcript uses "virus" to mean "bias")

The bias notation follows a similar logic to weights, but with **two subscripts** instead of three:

| Subscript | Meaning | Example |
|-----------|---------|---------|
| **i** | Which layer the bias belongs to | Layer 1, Layer 2, etc. |
| **j** | Which neuron in that layer the bias belongs to | Neuron 1, Neuron 2, etc. |

**So bias is written as: b_ij**

**Examples from the transcript context:**

```
If you have Layer 1 with 3 neurons:
- b_11 = bias for Layer 1, Neuron 1
- b_12 = bias for Layer 1, Neuron 2  
- b_13 = bias for Layer 1, Neuron 3

If you have Layer 2 with 2 neurons:
- b_21 = bias for Layer 2, Neuron 1
- b_22 = bias for Layer 2, Neuron 2
```

**Key rule from the transcript:**
- Input layer (Layer 0) has **NO biases** — only hidden layers and output layers have biases.

### Part B: Output Notation

**What is an output?**  
The output of a neuron is the number that comes out after applying the activation function to the weighted sum plus bias. This output then gets sent to neurons in the next layer.

**Output Notation from the Transcript**

The transcript states:

> *"The output of pretty much any note is exactly the same. The notation follows is 2 2"* (meaning "output" notation)

The transcript introduces output notation as: **o_ij**

| Subscript | Meaning | Example |
|-----------|---------|---------|
| **i** | Which layer the output comes from | Layer 1, Layer 2, etc. |
| **j** | Which neuron in that layer | Neuron 1, Neuron 2, etc. |

**So output is written as: o_ij**

**Examples:**

```
o_11 = output of Layer 1, Neuron 1
o_12 = output of Layer 1, Neuron 2
o_21 = output of Layer 2, Neuron 1
```

**Important concept from the transcript:**

> *"This output is the input for the next note."*

This means:
- **o_ij** (output of a neuron) becomes the **input** to neurons in the next layer.
- You will see outputs flowing forward through the network during **forward propagation**.

### How Outputs Connect to Weights

The transcript explains this connection clearly:

When you see a weight **w_ijk**:
- **j** refers to the **output of neuron j** from the previous layer
- That output (o_previousLayer_j) gets **multiplied** by the weight
- The result gets added to the input of neuron k in the current layer

**Visual representation:**

```
Previous Layer                    Current Layer
(Layer i-1)                       (Layer i)

o_(i-1)_1  ---(weight w_ij1)--->  (+) ---> neuron k=1
                                     \
o_(i-1)_2  ---(weight w_ij2)--->   (+) ---> neuron k=2
                                     /
o_(i-1)_3  ---(weight w_ij3)--->  (+)

Each output from previous layer connects to each neuron in current layer
with a unique weight.
```

---

## 3. Key Points

**Bias Notation:**
- Written as **b_ij** where:
  - i = layer number
  - j = neuron number within that layer
- Input layer (Layer 0) has **no biases**
- One bias per neuron in hidden and output layers

**Output Notation:**
- Written as **o_ij** where:
  - i = layer number
  - j = neuron number within that layer
- Output of one layer becomes **input** to the next layer

**Connection between outputs and weights:**
- Outputs flow forward through the network
- Each weight connects an output from previous layer to a neuron in current layer

## 7. Common Mistakes

| Mistake | Why it's wrong | How to avoid |
|---------|----------------|--------------|
| Giving biases to input layer neurons | Input neurons just pass data; they do not compute anything | Remember: only neurons that perform calculations (hidden + output) get biases |
| Confusing bias notation with output notation | Both use b_ij and o_ij — easy to mix up | b stands for **bias**, o stands for **output** (look at the letter) |
| Forgetting that outputs from one layer are inputs to the next | The notation hides this relationship | Always think of outputs as "signals traveling forward" |
| Using j to mean different things across notations | In weights, j = source neuron; in bias/output, j = neuron number within layer | Practice distinguishing the context of each notation |

## 8. Interview/Exam Questions

**Q1:** How do you write the bias for Neuron 3 in Layer 4 using the notation from the transcript?  
**A1:** b_43 (i=4, j=3)

**Q2:** Why does the input layer not have any biases?  
**A2:** The input layer's only job is to receive the raw data and pass it forward. It performs no mathematical operation (no weighted sum, no activation function), so it does not need a bias.

**Q3:** What is the relationship between o_(i-1)_j and w_ijk?  
**A3:** o_(i-1)_j is the output from neuron j in the previous layer. This output gets multiplied by weight w_ijk (where j is the same source neuron index) to contribute to the input of neuron k in the current layer i.

**Q4:** If a network has Layer 1 (4 neurons) and Layer 2 (3 neurons), how many biases are there in total?  
**A4:** 
- Layer 0 (input) has 0 biases
- Layer 1 (hidden) has 4 biases
- Layer 2 (output) has 3 biases
- Total biases = 4 + 3 = 7 biases
