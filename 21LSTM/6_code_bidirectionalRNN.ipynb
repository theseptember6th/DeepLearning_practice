```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,Bidirectional,SimpleRNN,LSTM,GRU,Dense
from tensorflow.keras import Input

#load the IMDB dataset
num_words=10000 #Number of words to keep
(X_train,y_train),(X_test,y_test)=imdb.load_data(num_words=num_words)

#pad sequences
maxlen=100 #maximum sequence length
X_train=pad_sequences(X_train,maxlen=maxlen,padding="post",truncating="post")
X_test=pad_sequences(X_test,maxlen=maxlen,padding="post",truncating="post")
```

### simple RNN model

```python
embedding_dim=32 # dimension of embedding layer

model=Sequential([
    Input(shape=(maxlen,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    SimpleRNN(units=5),
    Dense(units=1,activation="sigmoid") #binary classification (positive/negative)
])

#compile the model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])

#Display model architecture
model.summary()
```

### Now bidirectional RNN

```python


model=Sequential([
    Input(shape=(maxlen,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(SimpleRNN(units=5)), # two RNN forward and backward with each 5 RNN units
    Dense(units=1,activation="sigmoid") #binary classification (positive/negative)
])

#compile the model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])

#Display model architecture
model.summary()
```

### Bidirectional LSTM

```python


model=Sequential([
    Input(shape=(maxlen,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(LSTM(units=5)), # two RNN forward and backward with each 5 RNN units
    Dense(units=1,activation="sigmoid") #binary classification (positive/negative)
])

#compile the model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])

#Display model architecture
model.summary()
```

## Birdirectional GRU

```python


model=Sequential([
    Input(shape=(maxlen,)),
    Embedding(input_dim=num_words,output_dim=embedding_dim,input_length=maxlen),
    Bidirectional(GRU(units=5)), # two RNN forward and backward with each 5 RNN units
    Dense(units=1,activation="sigmoid") #binary classification (positive/negative)
])

#compile the model
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])

#Display model architecture
model.summary()
```
